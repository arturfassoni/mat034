# Método da Secante

O [método da secante](https://en.wikipedia.org/wiki/Secant_method) é muito semelhante ao método da bissecção, exceto que, em vez de dividir cada intervalo escolhendo o ponto médio, o método da secante divide cada intervalo pela reta secante que conecta os pontos extremos. O método da secante sempre converge para uma raiz de $f(x)=0$, desde que $f(x)$ seja contínua em $[a,b]$ e $f(a)f(b)<0$.

## Fórmula da Reta Secante

Seja $f(x)$ uma função contínua em um intervalo fechado $[a,b]$ tal que $f(a)f(b) < 0$. Uma solução da equação $f(x) = 0$ para $x \in [a,b]$ é garantida pelo [Teorema do Valor Intermediário](https://en.wikipedia.org/wiki/Intermediate_value_theorem). Considere a reta que conecta os valores dos pontos extremos $(a,f(a))$ e $(b,f(b))$. A reta que conecta esses dois pontos é chamada de reta secante e é dada pela fórmula

$$
y = \frac{f(b) - f(a)}{b - a}(x - a) + f(a)
$$

O ponto onde a reta secante cruza o eixo $x$ é

$$
0 = \frac{f(b) - f(a)}{b - a}(x - a) + f(a)
$$

o que resolvemos para $x$

$$
x = a - f(a)\frac{b - a}{f(b) - f(a)}
$$

## Algoritmo

O procedimento do método da secante é quase idêntico ao do método da bissecção. A única diferença é como dividimos cada subintervalo.

1. Escolha um intervalo inicial $[a_0,b_0]$ tal que $f(a_0)f(b_0) < 0$.
2. Calcule $f(x_0)$ onde $x_0$ é dado pela reta secante
  $$
  x_0 = a_0 - f(a_0)\frac{b_0 - a_0}{f(b_0) - f(a_0)}
  $$
3. Determine o próximo subintervalo $[a_1,b_1]$:
    1. Se $f(a_0)f(x_0) < 0$, então seja $[a_1,b_1]$ o próximo intervalo com $a_1=a_0$ e $b_1=x_0$.
    2. Se $f(b_0)f(x_0) < 0$, então seja $[a_1,b_1]$ o próximo intervalo com $a_1=x_0$ e $b_1=b_0$.
4. Repita (2) e (3) até que o intervalo $[a_N,b_N]$ atinja um comprimento predeterminado.
5. Retorne o valor $x_N$, a interceptação com o eixo $x$ do $N$-ésimo subintervalo.

Uma solução da equação $f(x)=0$ no intervalo $[a,b]$ é garantida pelo [Teorema do Valor Intermediário](https://en.wikipedia.org/wiki/Intermediate_value_theorem), desde que $f(x)$ seja contínua em $[a,b]$ e $f(a)f(b) < 0$. Em outras palavras, a função muda de sinal ao longo do intervalo e, portanto, deve ser igual a 0 em algum ponto no intervalo $[a,b]$.

## Implementação

Escreva uma função chamada `secant` que receba 4 parâmetros de entrada `f`, `a`, `b` e `N` e retorne a aproximação de uma solução de $f(x)=0$ dada por $N$ iterações do método da secante. Se $f(a_n)f(b_n) \geq 0$ em qualquer ponto da iteração (causado por um intervalo inicial ruim ou erro de arredondamento nos cálculos), então imprima `"Método da secante falha."` e retorne `None`.

```python
def secant(f,a,b,N):
    '''Approximate solution of f(x)=0 on interval [a,b] by the secant method.

    Parameters
    ----------
    f : function
        The function for which we are trying to approximate a solution f(x)=0.
    a,b : numbers
        The interval in which to search for a solution. The function returns
        None if f(a)*f(b) >= 0 since a solution is not guaranteed.
    N : (positive) integer
        The number of iterations to implement.

    Returns
    -------
    m_N : number
        The x intercept of the secant line on the the Nth interval
            m_n = a_n - f(a_n)*(b_n - a_n)/(f(b_n) - f(a_n))
        The initial interval [a_0,b_0] is given by [a,b]. If f(m_n) == 0
        for some intercept m_n then the function returns this solution.
        If all signs of values f(a_n), f(b_n) and f(m_n) are the same at any
        iterations, the secant method fails and return None.

    Examples
    --------
    >>> f = lambda x: x**2 - x - 1
    >>> secant(f,1,2,5)
    1.6180257510729614
    '''
    if f(a)*f(b) >= 0:
        print("Secant method fails.")
        return None
    a_n = a
    b_n = b
    for n in range(1,N+1):
        m_n = a_n - f(a_n)*(b_n - a_n)/(f(b_n) - f(a_n))
        f_m_n = f(m_n)
        if f(a_n)*f_m_n < 0:
            a_n = a_n
            b_n = m_n
        elif f(b_n)*f_m_n < 0:
            a_n = m_n
            b_n = b_n
        elif f_m_n == 0:
            print("Found exact solution.")
            return m_n
        else:
            print("Secant method fails.")
            return None
    return a_n - f(a_n)*(b_n - a_n)/(f(b_n) - f(a_n))
```

## Exemplos

Comparando com a bissecção:

```python
f = lambda x: x**2 - x - 1
approx_phi = bisection(f,1,2,25)
print(approx_phi)
print(abs(approx_phi-(1 + 5**0.5)/2))

approx_phi2 = secant(f,1,2,25)
print(approx_phi2)
print(abs(approx_phi2-(1 + 5**0.5)/2))

print(abs(approx_phi2-approx_phi))
```

### Razão Superáurea

Vamos testar nossa função com valores de entrada para os quais conhecemos a saída correta. Vamos encontrar uma aproximação da [razão superáurea](https://en.wikipedia.org/wiki/Supergolden_ratio): a única raiz real do polinômio $p(x) = x^3 - x^2 - 1$.

```python
p = lambda x: x**3 - x**2 - 1
print(p(1))
print(p(2))
x=np.linspace(-1,3,50)
plt.plot(x,p(x))
plt.plot(x,0*x)
plt.show()
```

Como o polinômio muda de sinal no intervalo $[1,2]$, podemos aplicar o método da secante com este como o intervalo inicial:

```python
approx = secant(p,1,2,20)
print(approx)
```

O valor exato da razão superáurea é

$$
\frac{1 + \sqrt[3]{\frac{29 + 3\sqrt{93}}{2}} + \sqrt[3]{\frac{29 - 3\sqrt{93}}{2}}}{3}
$$

```python
supergolden = (1 + ((29 + 3*93**0.5)/2)**(1/3) + ((29 - 3*93**0.5)/2)**(1/3))/3
print(supergolden)
```

Vamos comparar nossa aproximação com a solução exata:

```python
error = abs(supergolden - approx)
print(error)
```

Usando a bisecção:

```python
approx_bisec = bisection(p,1,2,20)
print(approx_bisec)
error2 = abs(supergolden - approx_bisec)
print(error2)
```
