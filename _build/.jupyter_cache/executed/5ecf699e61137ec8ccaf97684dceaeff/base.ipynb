```python
def add_row(A,k,i,j):
    """Soma k vezes a linha j à linha i na matriz A."""
    n = A.shape[0]
    E = np.eye(n)
    if i == j:
        E[i,i] = k + 1
    else:
        E[i,j] = k
    return E @ A
```

```python
M = np.array([[1,1],[3,2]])
print(M)
```

```python
add_row(M,2,0,1)
```

```python
add_row(M,3,1,1)
```

```python
def scale_row(A,k,i):
    """Multiplica a linha i por k na matriz A."""
    n = A.shape[0]
    E = np.eye(n)
    E[i,i] = k
    return E @ A
```

```python
M = np.array([[3,1],[-2,7]])
print(M)
```

```python
scale_row(M,3,1)
```

```python
A = np.array([[1,1,1],[1,-1,0]])
print(A)
```

```python
scale_row(A,5,1)
```

```python
def switch_rows(A,i,j):
    """Troca as linhas i e j na matriz A."""
    n = A.shape[0]
    E = np.eye(n)
    E[i,i] = 0
    E[j,j] = 0
    E[i,j] = 1
    E[j,i] = 1
    return E @ A
```

```python
A = np.array([[1,1,1],[1,-1,0]])
print(A)
```

```python
switch_rows(A,0,1)
```
