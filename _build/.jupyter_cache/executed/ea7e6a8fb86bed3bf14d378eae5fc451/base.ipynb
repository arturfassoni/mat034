```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.linalg as la
```

```python
A = np.array([[1,0],[0,-2]])
print(A)
```

```python
results = la.eig(A)
```

```python
print(results[0])
```

```python
print(results[1])
```

```python
eigvals, eigvecs = la.eig(A)
print(eigvals)
```

```python
print(eigvecs)
```

```python
eigvals = eigvals.real
print(eigvals)
```

```python
lambda1 = eigvals[1]
print(lambda1)
```

```python
v1 = eigvecs[:,1].reshape(2,1)
print(v1)
```

```python
A @ v1
```

```python
lambda1 * v1
```

```python
n = 4
P = np.random.randint(0,10,(n,n))
print(P)
```

```python
S = P @ P.T
print(S)
```

```python
evals, evecs = la.eig(S)
print(evals)
```

```python
evals = evals.real
print(evals)
```

```python
print(evecs)
```

```python
v1 = evecs[:,0] # A primeira coluna é o primeiro autovetor
print(v1)
```

```python
v2 = evecs[:,1] # A segunda coluna é o segundo autovetor
print(v2)
```

```python
v1 @ v2
```

```python
P = np.array([[1,1],[1,-1]])
print(P)
```

```python
D = np.diag((3,1))
print(D)
```

```python
M = P @ D @ la.inv(P)
print(M)
```

```python
evals, evecs = la.eig(M)
print(evals)
```

```python
print(evecs)
```

```python
Pinv = la.inv(P)
```

```python
k = 20
```

```python
%%timeit
result = M.copy()
for _ in range(1,k):
    result = result @ M
```

```python
%%timeit
P @ D**k @ Pinv
```
