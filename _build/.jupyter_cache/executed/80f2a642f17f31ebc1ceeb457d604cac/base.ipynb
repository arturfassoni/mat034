```python
def invertible(M):
    "Determine if the matrix M = [[a,b],[c,d]] is invertible."
    determinant = M[0][0] * M[1][1] - M[0][1] * M[1][0]
    if determinant != 0:
        return True
    else:
        return False
```

```python
invertible([[1,2],[3,4]])
```

```python
invertible([[1,1],[3,3]])
```

```python
def concave_up(p,a):
    "Determine if the polynomial p(x) is concave up at x=a."
    degree = len(p) - 1
    if degree < 2:
        return False
    else:
        # Compute the second derivative p''(a)
        DDp_a = sum([k*(k-1)*p[k]*a**(k-2) for k in range(2,degree + 1)])
        if DDp_a > 0:
            return True
        else:
            return False
```

```python
p = [1,1,0,-1]
a = 2
concavity = concave_up(p,a)
print(concavity)
```
