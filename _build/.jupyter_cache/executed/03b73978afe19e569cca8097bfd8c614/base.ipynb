```python
N = np.random.randint(0,10,[2,2])
print(N)
```

```python
trace_N = np.trace(N)
det_N = la.det(N)
I = np.eye(2)
N @ N - trace_N * N + det_N * I
```

```python
v = np.array([1,2,3])
w = np.array([1,1,1])
print(v*w)
print(np.sum(v*w))
```

```python
def proj(v,w):
    '''Projeta o vetor v em w.'''
    v = np.array(v)
    w = np.array(w)
    return np.sum(v * w)/np.sum(w * w) * w   # ou (v @ w)/(w @ w) * w
```

```python
proj([1,2,3],[1,1,1])
```
