# Matplotlib

[Matplotlib](http://matplotlib.org) é um pacote Python para plotagem 2D e o submódulo `matplotlib.pyplot` contém muitas funções de plotagem para criar vários tipos de gráficos. Vamos começar importando `matplotlib.pyplot`.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Plotagem Básica

### Procedimento

O procedimento geral para criar um gráfico de linha 2D é:

1. Crie uma sequência de valores $x$.
2. Crie uma sequência de valores $y$.
3. Digite `plt.plot(x,y,[fmt],**kwargs)` onde `[fmt]` é uma string de formatação (opcional) e `**kwargs` são argumentos de palavra-chave (opcionais) que especificam as propriedades da linha do gráfico.
4. Use as funções do `pyplot` para adicionar recursos à figura, como título, legenda, linhas de grade, etc.
5. Digite `plt.show()` para exibir a figura resultante.

Vamos começar com um exemplo básico com alguns pontos aleatórios:

```python
x = [-5,-2,0,1,3]
y = [2,-1,1,-4,3]
plt.plot(x,y)
plt.show()
```

Os principais pontos a notar são:

1. As sequências `x` e `y` definem as coordenadas dos pontos no gráfico.
2. A linha no gráfico é construída conectando os pontos por linhas retas.

A segunda observação implica que, se quisermos plotar uma curva suave, precisamos plotar muitos pontos, caso contrário, o gráfico não será suave. Por exemplo, poderíamos tentar plotar a parábola $y = x^2$ para $x \in [-2,2]$ usando apenas 5 pontos:

```python
x = [-2,-1,0,1,2]
y = [4,1,0,1,4]
plt.plot(x,y)
plt.show()
```

Isso são poucos pontos para plotar uma curva suave como $y = x^2$ e, portanto, precisamos de mais pontos! Vamos tentar novamente usando a função do NumPy `np.linspace` para criar 100 pontos!

```python
x = np.linspace(-2,2,100+1)
y = x**2
plt.plot(x,y)
plt.show()
```

Essa é uma representação melhor da parábola $y = x^2$. Note que o número de pontos que usamos em um gráfico de linha (100 neste caso) é completamente arbitrário, mas o objetivo é mostrar um gráfico suave para uma curva suave e, portanto, só precisamos escolher um número grande o suficiente, dependendo da função. Mas tenha cuidado para não gerar muitos pontos, pois um número *muito* grande de pontos levará *muito* tempo para plotar!

Agora que temos a ideia geral, vamos ver como adicionar estilo e recursos aos nossos gráficos!

```python
x = np.linspace(-np.pi,np.pi,100+1)
y = np.sin(x)
plt.plot(x,y)
plt.show()
```

### Propriedades da Linha

Uma linha que aparece em um gráfico tem várias propriedades, incluindo cor, transparência, estilo, largura e marcadores. Podemos definir essas propriedades quando chamamos `plt.plot` usando os seguintes argumentos de palavra-chave:

| Propriedade | Descrição |
| :---: | :--- |
| `alpha` | transparência (0.0 transparente a 1.0 opaco) |
| `color` (ou `c`) | qualquer cor do matplotlib |
| `label` | texto que aparece na legenda |
| `linestyle` (ou `ls`) | `solid`, `dashed`, `dashdot`, `dotted` |
| `linewidth` (ou `lw`) | define a largura da linha |
| `marker` | define o estilo do marcador |
| `markeredgecolor` (ou `mec`) | qualquer cor do matplotlib |
| `markerfacecolor` (ou `mfc`) | qualquer cor do matplotlib |
| `markersize` (ou `ms`) | tamanho do marcador |

Observe que podemos especificar uma [cor do matplotlib](https://matplotlib.org/api/colors_api.html) de várias maneiras diferentes, incluindo pelo nome, como `blue` ou `red`, ou por uma tupla [RGB](https://www.w3schools.com/colors/colors_rgb.asp), como `(1,0,1)` para roxo. Por exemplo, vamos plotar a função

$$
y = e^{-x^2}\cos(2 \pi x) \ \ , \ \ x \in [-2,2]
$$

```python
x_inicial = -2
x_final = 2
n_pontos = 80
x = np.linspace(x_inicial,x_final,n_pontos+1)
y = np.exp(-x**2) * np.cos(2*np.pi*x)
plt.plot(x,y
         ,alpha=0.4,label='Decaying Cosine',
         color='red',linestyle='dashed',linewidth=2,
         marker='v',markersize=5,markerfacecolor='blue',
         markeredgecolor='green'
         )
plt.ylim([-2,2])
plt.legend()
plt.show()
```

Note que usamos a função do pyplot `plt.legend` para exibir a figura com uma legenda (mostrando o rótulo da linha) e `plt.ylim` para definir os limites no eixo vertical para `[-2,2]`.

### Strings de Formatação

Uma [string de formatação](https://matplotlib.org/api/_as_gen/matplotlib.pyplot.plot.html#matplotlib-pyplot-plot) nos dá um atalho para adicionar cor, marcadores e estilo de linha a um gráfico de linha. Por exemplo, se quisermos plotar a função

$$
y = \frac{1}{1 + x^2} \ , \ x \in [-5,5]
$$

com uma linha preta tracejada e marcadores quadrados, poderíamos usar argumentos de palavra-chave:

```python
x = np.linspace(-5,5,41)
y = 1/(1 + x**2)
plt.plot(x,y,color='black',linestyle='dashed',marker='s',linewidth=3)
plt.show()
```

Ou poderíamos usar a string de formatação correspondente `'ks--'` onde `k` denota uma linha preta, `s` um marcador quadrado e `--` uma linha tracejada:

```python
x = np.linspace(-5,5,41)
y = 1/(1 + x**2)
plt.plot(x,y,'ks--',lw=3)
plt.show()
```

### Polinômios de Taylor

Plote a função $y = \cos(x)$ juntamente com seus [polinômios de Taylor](https://en.wikipedia.org/wiki/Taylor_series) de graus 2 e 4.

```python
x = np.linspace(-6,6,50)

# Plot y = cos(x)
y = np.cos(x)
plt.plot(x,y,'b',label='cos(x)')

# Plot degree 2 Taylor polynomial
y2 = 1 - x**2/2
plt.plot(x,y2,'r-.',label='Degree 2')

# Plot degree 4 Taylor polynomial
y4 = 1 - x**2/2 + x**4/24
plt.plot(x,y4,'g:',label='Degree 4')

# Add features to our figure
plt.legend()
plt.grid(True,linestyle=':',alpha=0.3)
plt.yticks(np.arange(-6,6.1,0.5))
plt.xlim([-6,6])
plt.ylim([-3,3])
plt.title('Taylor Polynomials of cos(x) at x=0')
plt.xlabel('eixo x')
plt.ylabel('eixo y')

plt.show()
```

\begin{align}
x &= 16 \sin^3(t) \\\
y &= 13 \cos(t) - 5 \cos(2t) - 2 \cos(3t) - \cos(4t)

para $t \in [0,2\pi]$.

## Subplots (Gráficos Múltiplos)

A função `plt.subplot` recebe pelo menos 3 entradas $n$, $m$ e $i$ e cria uma figura com uma grade de $n$ por $m$ subplots e, em seguida, define o $i$-ésimo subplot (contando ao longo das linhas) como o gráfico atual (ou seja, o objeto de eixos atual).

Por exemplo, considere a [onda dente de serra](https://en.wikipedia.org/wiki/Sawtooth_wave)

$$
f(t) = \frac{1}{2} - \frac{1}{\pi} \sum_{k=1}^{\infty} (-1)^k \frac{\sin(2 \pi k t)}{k}
$$

e seja $f_N(t)$ a $N$-ésima soma parcial da onda dente de serra:

$$
f_N(t) = \frac{1}{2} - \frac{1}{\pi} \sum_{k=1}^{N} (-1)^k \frac{\sin(2 \pi k t)}{k}
$$

Crie uma grade de 2 por 2 de subplots para plotar as 4 primeiras somas parciais:

\begin{align}
f_1(t) &= \frac{1}{2} + \frac{\sin(2 \pi t)}{\pi} \\\
f_2(t) &= \frac{1}{2} + \frac{\sin(2 \pi t)}{\pi} - \frac{\sin(4 \pi t)}{2\pi} \\\
f_3(t) &= \frac{1}{2} + \frac{\sin(2 \pi t)}{\pi} - \frac{\sin(4 \pi t)}{2\pi} + \frac{\sin(6 \pi t)}{3\pi} \\\
f_4(t) &= \frac{1}{2} + \frac{\sin(2 \pi t)}{\pi} - \frac{\sin(4 \pi t)}{2\pi} + \frac{\sin(6 \pi t)}{3\pi} - \frac{\sin(8 \pi t)}{4\pi}
\end{align}

```python
t = np.linspace(0,4,200)

fN = 1/2
for k in [1,2,3,4,5,6]:
    fN = fN - (-1)**k * np.sin(2*k*np.pi*t)/(k*np.pi)
    plt.subplot(3,2,k)
    plt.plot(t,fN)
    plt.title('N = {}'.format(k))

plt.tight_layout()
plt.show()
```

Veja a [documentação](https://matplotlib.org/api/_as_gen/matplotlib.pyplot.subplot.html) para mais informações sobre subplots.

### Gráficos de Dispersão (Scatter plots)

Um [gráfico de dispersão](https://matplotlib.org/api/_as_gen/matplotlib.pyplot.scatter.html) tem 4 dimensões: coordenada $x$, coordenada $y$, tamanho e cor. Vamos fazer um gráfico de dispersão aleatório:

```python
# Set the number of dots in the plot
N = 2000

# Create random x and y coordinates sampled uniformly from [0,1]
x = np.random.rand(N)
y = np.random.rand(N)

# Create random array sampled uniformly from [20,120]
# `size` array is used below to set the size of each dot
size = 100*np.random.rand(N) + 20

# Create random 4-tuples sampled uniformly from [0,1]
# `colors` array is used below to set the color
# (red,green,blue,alpha) of each dot
colors = np.random.rand(N,4)

# Create a figure of size 12 by 5 and create scatter plot
plt.figure(figsize=(7,5))
plt.scatter(x,y,c=colors,s=size)
plt.axis('off')
plt.show()
```

## Objetos Figure e Axes

Até agora, usamos o `pyplot` (`plt`) para criar gráficos de forma "automática". Funções como `plt.plot()` e `plt.title()` atuam na figura ou no gráfico que estiver "ativo" no momento. Essa é a **interface baseada em estado** (state-based) do Matplotlib. Ela é ótima para gráficos rápidos e simples.

No entanto, para ter mais controle, especialmente ao criar múltiplos gráficos em uma única figura, usamos a **interface orientada a objetos**. Essa abordagem é mais explícita e robusta.

Os dois objetos centrais são:

* **`Figure`**: É o contêiner de mais alto nível, a "tela" ou a janela inteira onde tudo é desenhado. Uma `figure` pode conter um ou mais objetos `Axes`.
* **`Axes`**: É o gráfico em si, a área onde os dados são plotados com eixos x e y, títulos, legendas, etc. **Não confunda `Axes` (o objeto) com `axis` (eixo, como x ou y)**. Um objeto `Axes` contém dois (ou três, em 3D) objetos `Axis`.

A maneira mais comum de começar é com a função `plt.subplots()`. Ela cria uma figura e um conjunto de subplots (os objetos `Axes`) de uma só vez.

```python
# Criar os objetos Figure e Axes
# fig é a tela inteira, ax é o nosso gráfico
fig, ax = plt.subplots()

# Agora usamos os métodos do objeto 'ax' para plotar
x = np.linspace(0, 10, 200)
y = np.sin(x) * np.exp(-0.1 * x)

ax.plot(x, y)

plt.show()
```

Note que, em vez de `plt.funcao()`, agora chamamos `ax.metodo()`. A sintaxe é um pouco diferente, mas muito mais clara. Muitos dos métodos para configurar o gráfico usam o prefixo `set_`.

| Interface `pyplot` (baseada em estado) | Interface Orientada a Objetos (em um objeto `ax`) |
| :--- | :--- |
| `plt.plot(x, y)` | `ax.plot(x, y)` |
| `plt.title('Título')` | `ax.set_title('Título')` |
| `plt.xlabel('Eixo X')` | `ax.set_xlabel('Eixo X')` |
| `plt.ylabel('Eixo Y')` | `ax.set_ylabel('Eixo Y')` |
| `plt.grid(True)` | `ax.grid(True)` |
| `plt.legend()` | `ax.legend()` |
| `plt.xlim([0, 10])` | `ax.set_xlim([0, 10])` |

A verdadeira vantagem da abordagem orientada a objetos se torna clara ao trabalhar com múltiplos subplots em uma única figura.

```python
# Criar uma figura com 1 linha e 2 colunas de subplots
# axs agora é um array de objetos Axes
fig, axs = plt.subplots(1,2, figsize=(12, 5)) # figsize controla o tamanho total da figura

# ---------- Gráfico 1 (à esquerda) ----------
x1 = np.linspace(0, 2 * np.pi, 100)
y1 = np.cos(x1)
axs[0].plot(x1, y1, color='blue',linestyle='--',label='cos x')
axs[0].set_title('Gráfico de Cosseno')
axs[0].set_xlabel('Ângulo [rad]')
axs[0].set_ylabel('Amplitude')
axs[0].grid(True,alpha=0.2,c='gray')
axs[0].legend()


# ---------- Gráfico 2 (à direita) ----------
x2 = np.linspace(-3, 3, 100)
y2 = np.exp(-x2**2)
axs[1].plot(x2, y2, color='red')
axs[1].set_title('Curva Gaussiana')
axs[1].set_xlabel('x')
# O ylabel do segundo gráfico pode ser omitido para um visual mais limpo
axs[1].grid(True,alpha=0.1)

# Ajusta o layout para evitar sobreposição de títulos/rótulos
plt.tight_layout()

plt.show()
```

Com essa abordagem, temos controle total sobre cada subplot individualmente, acessando-os pelo índice `axs[0]`, `axs[1]`, etc. Isso torna a criação de dashboards e visualizações complexas muito mais organizada e gerenciável.
