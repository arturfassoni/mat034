# Laços

## Laços for

Um [laço for](https://docs.python.org/3/reference/compound_stmts.html#for) nos permite executar um bloco de código várias vezes com alguns parâmetros atualizados a cada iteração do laço. Um laço `for` começa com a declaração `for`:

```python
iterable = [1,2,3,4]
for item in iterable:
    # bloco de código indentado com 4 espaços
    print(item)
```

Os principais pontos a serem observados são:

* As palavras-chave `for` e `in`
* `iterable` é um objeto de sequência, como uma lista, tupla ou range
* `item` é uma variável que assume cada valor em `iterable`
* termine a declaração `for` com dois pontos `:`
* o bloco de código é indentado com 4 espaços e executa uma vez para cada valor em `iterable`

Por exemplo, vamos imprimir $n^2$ para $n$ de 0 a 5:

```python
for n in range(0,5+1):#[0,1,2,3,4,5]:
    square = n**2
    print(n,'squared is',square)
print('The for loop is complete!')
```

Copie e cole este código e qualquer um dos exemplos abaixo no [visualizador Python](http://www.pythontutor.com/visualize.html#mode=edit) para ver cada passo em um laço `for`!

## Laços while

E se quisermos executar um bloco de código várias vezes, mas não sabemos exatamente quantas vezes? Não podemos escrever um laço `for` porque isso exige que definamos o comprimento do laço com antecedência. Esta é uma situação em que um [laço while](https://en.wikipedia.org/wiki/While_loop#Python) é útil.

O exemplo a seguir ilustra um [laço while](https://docs.python.org/3/tutorial/introduction.html#first-steps-towards-programming):

```python
n = 5
while n > 0:
    print(n)
    n = n - 1
```

Os principais pontos a serem observados são:

* A palavra-chave `while`
* uma expressão lógica seguida por dois pontos `:`
* o laço executa seu bloco de código se a expressão lógica for avaliada como `True`
* atualize a variável na expressão lógica a cada iteração do laço
* **CUIDADO!** Se a expressão lógica *sempre* for avaliada como `True`, você terá um [laço infinito](https://en.wikipedia.org/wiki/While_loop#Python)!

Preferimos laços `for` a laços `while` por causa do último ponto. Um laço `for` nunca resultará em um laço infinito. Se um laço pode ser construído com `for` ou `while`, sempre escolheremos `for`.

## Construindo Sequências

Existem várias maneiras de construir uma sequência de valores e salvá-los como uma lista Python. Já vimos a sintaxe de list comprehension do Python. Há também o método de lista `append`, descrito abaixo.

### Sequências por uma Fórmula

Se uma sequência é dada por uma fórmula, podemos usar uma list comprehension para construí-la. Por exemplo, a sequência de quadrados de 1 a 100 pode ser construída usando uma list comprehension:

```python
squares = [d**2 for d in range(1,11)]
print(squares)
```

No entanto, podemos alcançar o mesmo resultado com um laço `for` e o método `append` para listas:

```python
# Inicializa uma lista vazia
squares = []
for d in range(1,11):
    # Anexa o próximo quadrado à lista
    squares.append(d**2)
print(squares)
```

Na verdade, os dois exemplos acima são equivalentes. O propósito das list comprehensions é simplificar e comprimir a sintaxe em uma construção de uma linha.

### Sequências Recursivas

Só podemos usar uma list comprehension para construir uma sequência quando os valores da sequência são definidos por uma fórmula. Mas e se quisermos construir uma sequência onde o próximo valor depende dos valores anteriores? Isso é chamado de [sequência recursiva](https://en.wikipedia.org/wiki/Recursion).

Por exemplo, considere a [sequência de Fibonacci](https://en.wikipedia.org/wiki/Fibonacci_number):

$$
x_1 = 1, x_2 = 1, x_3 = 2, x_4 = 3, x_5 = 5, ...
$$

onde

$$
x_{n} = x_{n-1} + x_{n-2}
$$

Não podemos usar uma list comprehension para construir a lista de números de Fibonacci, então devemos usar um laço `for` com o método `append`. Por exemplo, os primeiros 15 números de Fibonacci são:

```python
fibonacci_numbers = [1,1]
for n in range(2,15):
    fibonacci_n = fibonacci_numbers[n-1] + fibonacci_numbers[n-2]
    fibonacci_numbers.append(fibonacci_n)
    print(fibonacci_numbers)
```

## Calculando Somas

Suponha que queremos calcular a soma de uma sequência de números $x_0$, $x_1$, $x_2$, $x_3$, $\dots$, $x_n$. Existem pelo menos duas abordagens:

1. Calcule a sequência inteira, armazene-a como uma lista $[x_0,x_1,x_2,\dots,x_n]$ e, em seguida, use a função embutida `sum`.
2. Inicialize uma variável com o valor 0 (e nomeie-a `resultado`, por exemplo), crie e adicione cada elemento da sequência a `resultado`, um de cada vez.

A vantagem da segunda abordagem é que não precisamos armazenar todos os valores de uma vez. Por exemplo, aqui estão duas maneiras de escrever uma função que calcula a soma dos quadrados.

Para a primeira abordagem, use uma list comprehension:

```python
def sum_of_squares_1(N):
    "Compute the sum of squares 1**2 + 2**2 + ... + N**2."
    return sum([n**2 for n in range(1,N + 1)])
```

```python
sum_of_squares_1(4)
```

Para a segunda abordagem, use um laço `for` com a construção de inicialização e atualização:

```python
def sum_of_squares_2(N):
    "Compute the sum of squares 1**2 + 2**2 + ... + N**2."
    # Inicializa o valor de saída para 0
    result = 0
    for n in range(1,N + 1):
        # Atualiza o resultado adicionando o próximo termo
        result = result + n**2
        print(result)
    return result
```

```python
sum_of_squares_2(4)
```

Novamente, ambos os métodos produzem o mesmo resultado, no entanto, o segundo usa menos memória!

## Calculando Produtos

Não há uma função embutida para calcular produtos de sequências, portanto, usaremos uma construção de inicialização e atualização semelhante ao exemplo acima para calcular somas.

Escreva uma função chamada `factorial` que receba um inteiro positivo $N$ e retorne o fatorial $N!$.

```python
def factorial(N):
    "Compute N! = N(N-1) ... (2)(1) for N >= 1."
    # Inicializa a variável de saída para 1
    product = 1
    for n in range(2,N + 1):
        # Atualiza a variável de saída, note que N!=N(N-1)!
        product = product * n
    return product
```

Vamos testar nossa função para valores de entrada cujo resultado conhecemos:

```python
factorial(2)
```

```python
factorial(15)
```

Podemos usar nossa função para aproximar $e$ usando a série de Taylor para $e^x$:

$$
e^x = \sum_{k=0}^{\infty} \frac{x^k}{k!}
$$

Por exemplo, vamos calcular a 100ª soma parcial da série com $x=1$:

```python
sum([1/factorial(k) for k in range(0,101)])
```

## Procurando por Soluções

Podemos usar laços `for` para procurar soluções inteiras de equações. Por exemplo, suponha que gostaríamos de encontrar todas as representações de um inteiro positivo $N$ como uma [soma de dois quadrados](https://en.wikipedia.org/wiki/Sum_of_two_squares_theorem). Em outras palavras, queremos encontrar todas as soluções inteiras $(x,y)$ da equação:

$$
x^2 + y^2 = N
$$

Escreva uma função chamada `reps_sum_squares` que receba um inteiro $N$ e encontre todas as representações de $N$ como uma soma de quadrados $x^2 + y^2 = N$ para $0 \leq x \leq y$. A função retorna as representações como uma lista de tuplas. Por exemplo, se $N = 50$, então $1^2 + 7^2 = 50$ e $5^2 + 5^2 = 50$, e a função retorna a lista `[(1, 7),(5, 5)]`.

Vamos delinear nossa abordagem antes de escrever qualquer código:

1. Dado $x \leq y$, o maior valor possível para $x$ é $\sqrt{\frac{N}{2}}$
2. Para $x \leq \sqrt{\frac{N}{2}}$, o par $(x,y)$ é uma solução se $N - x^2$ for um quadrado
3. Defina uma função auxiliar chamada `is_square` para testar se um inteiro é um quadrado

```python
N=72
xmax=(N/2)**0.5
print(xmax)
int(xmax)
```

```python
def is_square(n):
    "Determine if the integer n is a square."
    if round(n**0.5)**2 == n:
        return True
    else:
        return False

def reps_sum_squares(N):
    '''Find all representations of N as a sum of squares x**2 + y**2 = N.

    Parameters
    ----------
    N : integer

    Returns
    -------
    reps : list of tuples of integers
        List of tuples (x,y) of positive integers such that x**2 + y**2 = N.

    Examples
    --------
    >>> reps_sum_squares(1105)
    [(4, 33), (9, 32), (12, 31), (23, 24)]
    '''
    reps = []
    if is_square(N/2):
        # If N/2 is a square, search up to x = (N/2)**0.5
        max_x = round((N/2)**0.5)
    else:
        # If N/2 is not a square, search up to x = floor((N/2)**0.5)
        max_x = int((N/2)**0.5)
    for x in range(0,max_x + 1):
        y_squared = N - x**2
        if is_square(y_squared):
            y = round(y_squared**0.5)
            # Append solution (x,y) to list of solutions
            reps.append((x,y))
    return reps
```

```python
out = reps_sum_squares(1105)
print(out)
```

```python
soma_quadrados = lambda x,y: x**2+y**2
```

```python
soma_quadrados(4,33)
```

```python
for i in range(0,len(out)):
  print("para o par (x,y)=",out[i],"a soma x^2+yˆ2 é",soma_quadrados(out[i][0],out[i][1]))
```

### Números Primos

Um inteiro positivo é [primo](https://en.wikipedia.org/wiki/Prime_number) se for divisível apenas por 1 e por ele mesmo. Escreva uma função chamada `is_prime` que receba um parâmetro de entrada `n` e retorne `True` ou `False` dependendo se `n` é primo ou não.

Vamos delinear nossa abordagem antes de escrever qualquer código:

1. Um inteiro $d$ divide $n$ se não houver resto na divisão de $n$ por $d$.
2. Use o operador de módulo `%` para calcular o resto.
3. Se $d$ divide $n$, então $n = d q$ para algum inteiro $q$ e ou $d \leq \sqrt{n}$ ou $q \leq \sqrt{n}$ (e não ambos), portanto, só precisamos testar se $d$ divide $n$ para inteiros $d \leq \sqrt{n}$

```python
def is_prime(n):
    "Determine whether or not n is a prime number."
    if n <= 1:
        return False
    # Test if d divides n for d <= n**0.5
    for d in range(2,round(n**0.5) + 1):
        if n % d == 0:
            # n is divisible by d and so n is not prime
            return False
    # If we exit the for loop, then n is not divisible by any d
    # and therefore n is prime
    return True
```

Vamos testar nossa função nos primeiros 30 números:

```python
for n in range(0,31):
    if is_prime(n):
        print(n,'is prime!')
```

Nossa função funciona! Vamos encontrar todos os primos entre 20.000 e 20.100.

```python
for n in range(20000,20100):
    if is_prime(n):
        print(n,'is prime!')
```

### Divisores

Vamos escrever uma função chamada `divisors` que recebe um inteiro positivo $N$ e retorna a lista de inteiros positivos que dividem $N$.

```python
15//2
```

```python
def divisors(N):
    "Return the list of divisors of N."
    # Initialize the list of divisors (which always includes 1)
    divisor_list = [1]
    # Check division by d for d <= N/2
    for d in range(2,N // 2 + 1):
        if N % d == 0:
            divisor_list.append(d)
    # N divides itself and so we append N to the list of divisors
    divisor_list.append(N)
    return divisor_list
```

### Conjectura de Collatz

Seja $a$ um inteiro positivo e considere a sequência recursiva onde $x_0 = a$ e

$$
x_{n+1} = \left\{ \begin{array}{cl} x_n/2 & \text{se } x_n \text{ for par} \\\\ 3x_n+1 & \text{se } x_n \text{ for ímpar}  \end{array} \right.
$$

A [Conjectura de Collatz](https://en.wikipedia.org/wiki/Collatz_conjecture) afirma que esta sequência *sempre* alcançará 1. Por exemplo, se $a = 10$, então $x_0 = 10$, $x_1 = 5$, $x_2 = 16$, $x_3 = 8$, $x_4 = 4$, $x_5 = 2$ e $x_6 = 1$.

Escreva uma função chamada `collatz` que receba um parâmetro de entrada `a` e retorne a sequência de inteiros definida acima e terminando com a primeira ocorrência de $x_n=1$.

```python
def collatz(a):
    "Compute the Collatz sequence starting at a and ending at 1."
    # Initialize list with first value a
    sequence = [a]
    # Compute values until we reach 1
    while sequence[-1] > 1: # while the last term is > 1
        # Check if the last element in the list is even
        if sequence[-1] % 2 == 0:
            # Compute and append the new value
            sequence.append(sequence[-1] // 2)
        else:
            # Compute and append the new value
            sequence.append(3*sequence[-1] + 1)
    return sequence
```

Vamos testar nossa função:

```python
print(collatz(10))
```

```python
collatz(43)
```

A conjectura de Collatz é incrível. Não importa onde comecemos, a sequência sempre termina em 1!

```python
a = 123456789
seq = collatz(a)
print("Collatz sequence for a =",a)
print("begins with",seq[:5])
print("ends with",seq[-5:])
print("and has",len(seq),"terms.")
```

Qual $a < 1000$ produz a sequência mais longa?

```python
max_length = 1
a_max = 1
for a in range(1,1001):
    seq_length = len(collatz(a))
    if seq_length > max_length:
        max_length = seq_length
        a_max = a
        seq_max = collatz(a)
print('Longest sequence begins with a =',a_max,'and has length',max_length)
print('The last 15 terms are',seq_max[-15:])
```
