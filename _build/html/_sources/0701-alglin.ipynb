## Arrays NumPy

Vamos começar com uma rápida revisão dos [Arrays NumPy](../../scipy/numpy/). Podemos pensar em um array NumPy 1D como uma lista de números. Podemos pensar em um array NumPy 2D como uma matriz. E podemos pensar em um array 3D como um cubo de números.
Quando selecionamos uma linha ou coluna de um array NumPy 2D, o resultado é um array NumPy 1D (chamado de [fatia ou slice](https://docs.scipy.org/doc/numpy/reference/arrays.indexing.html#basic-slicing-and-indexing)). Isso é diferente do MATLAB, onde, ao selecionar uma coluna de uma matriz, ela é retornada como um vetor coluna, que é uma matriz 2D do MATLAB.

Isso pode ficar um pouco confuso, então precisamos ficar atentos à forma (shape), tamanho (size) e dimensão dos nossos arrays NumPy.

### Atributos do Array

Crie um array NumPy 1D (unidimensional) e verifique suas dimensões, forma e tamanho.

Crie um array NumPy 2D (bidimensional) (ou seja, uma matriz):

Verifique o número de dimensões da fatia:

```python
col.ndim
```

Quando selecionamos uma linha ou coluna de um array NumPy 2D, o resultado é um array NumPy 1D. No entanto, podemos querer selecionar uma coluna como um vetor coluna 2D. Isso exige que usemos o método [reshape](https://docs.scipy.org/doc/numpy/reference/generated/numpy.reshape.html).

### Polinômios Característicos e Teorema de Cayley-Hamilton

O polinômio característico de uma matriz quadrada $A$ 2x2 é

$$
p_A(\lambda) = \det(A - \lambda I) = \lambda^2 - \mathrm{tr}(A) \lambda + \mathrm{det}(A)
$$

O [Teorema de Cayley-Hamilton](https://en.wikipedia.org/wiki/Cayley%E2%80%93Hamilton_theorem) afirma que toda matriz quadrada satisfaz seu polinômio característico. Para uma matriz $A$ de tamanho 2, isso significa que

$$
p_A(A) = A^2 - \mathrm{tr}(A) A + \mathrm{det}(A) I = 0
$$

Vamos verificar o Teorema de Cayley-Hamilton para algumas matrizes diferentes.

```python
print(A)
```

```python
trace_A = np.trace(A)
det_A = la.det(A)
I = np.eye(2)
A @ A - trace_A * A + det_A * I
```

Vamos fazer isso novamente para algumas matrizes aleatórias:

```python
N = np.random.randint(0,10,[2,2])
print(N)
```

```python
trace_N = np.trace(N)
det_N = la.det(N)
I = np.eye(2)
N @ N - trace_N * N + det_N * I
```

### Projeções

A fórmula para projetar um vetor $v$ em um vetor $w$ é

$$
\mathrm{proj}_w(v) = \frac{v \cdot w}{w \cdot w} w
$$

Vamos escrever uma função chamada `proj` que calcula a projeção de $v$ em $w$.

Para isso, note que o produto interto $v \cdot w$ pode ser calculado como:

```python
v = np.array([1,2,3])
w = np.array([1,1,1])
print(v*w)
print(np.sum(v*w))
```

```python
def proj(v,w):
    '''Projeta o vetor v em w.'''
    v = np.array(v)
    w = np.array(w)
    return np.sum(v * w)/np.sum(w * w) * w   # ou (v @ w)/(w @ w) * w
```

```python
proj([1,2,3],[1,1,1])
```
