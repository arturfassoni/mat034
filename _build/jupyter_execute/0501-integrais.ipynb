# Integrais Definidas

A [integral definida](https://en.wikipedia.org/wiki/Integral) de uma função $f(x)$ sobre um intervalo $[a,b]$ é o limite

$$
\int_a^b f(x) \, dx = \lim_{N \to \infty} \sum_{i=1}^N f(x_i^ * ) (x_i - x_{i-1}) \ \ , \ x_i^* \in [x_{i-1},x_i]
$$

onde, para cada $N$,

$$
x_0 = a < x_1 < \cdots < x_N = b
$$

é uma partição de $[a,b]$ com $N$ subintervalos e os valores $x_i^ * \in [x_{i-1},x_i]$ escolhidos em cada subintervalo são arbitrários.

A integral definida representa a área (líquida) sob a curva do gráfico de $y=f(x)$ no intervalo $[a,b]$.

$$
\int_a^b f(x) \, dx = \text{(área líquida) sob a curva } y = f(x) \text{ em } [a,b]
$$

O termo "líquida" significa que a área acima do eixo $x$ é positiva e a área abaixo do eixo $x$ conta como área negativa. Por exemplo, podemos visualizar a integral:

$$
\int_{\pi/2}^{3\pi/2} \left( \sin(0.2 x) + \sin(2x) + 1 \right) dx
$$

```python
import numpy as np
import matplotlib.pyplot as plt

f = lambda x: np.sin(0.2*x) + np.sin(2*x) + 1

x = np.linspace(0,2*np.pi,100)
y = f(x)
plt.plot(x,y)

X = np.linspace(np.pi/2,3*np.pi/2,100)
Y = f(X)
plt.fill_between(X,Y)

plt.xticks([np.pi/2,3*np.pi/2],['$\pi/2$','$3\pi/2$']); plt.yticks([-1,0,1,2,3]);
plt.xlim([0,2*np.pi]); plt.ylim([0,3]);
plt.show()
```

Em nossos cursos introdutórios de cálculo, focamos em integrais que podemos resolver exatamente pelo [Teorema Fundamental do Cálculo](https://en.wikipedia.org/wiki/Fundamental_theorem_of_calculus), como

$$
\int_0^{\pi/2} \cos(x) \, dx = \sin(\pi/2) - \sin(0) = 1
$$

No entanto, a maioria das integrais definidas é impossível de resolver exatamente. Por exemplo, a famosa [função erro](https://en.wikipedia.org/wiki/Error_function) em probabilidade

$$
\mathrm{erf}(x) = \frac{2}{\sqrt{\pi}} \int_0^x e^{-t^2} dt
$$

é uma integral definida para a qual não existe fórmula explícita.

A ideia por trás da [integração numérica](https://en.wikipedia.org/wiki/Numerical_integration) é usar formas geométricas simples para aproximar a área sob a curva $y=f(x)$ para estimar integrais definidas. Nesta seção, exploramos os métodos mais simples de integração numérica: somas de Riemann, a regra do trapézio e a regra de Simpson.
