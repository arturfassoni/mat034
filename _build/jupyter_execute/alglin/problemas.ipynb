**Exercício 3.** Análise de Circuito Elétrico (Sistemas Lineares)**

A Lei das Malhas de Kirchhoff afirma que a soma das quedas de tensão em qualquer malha fechada de um circuito é igual à tensão total fornecida pelas fontes de energia nessa malha. Para um resistor, a queda de tensão é $V = IR$. Isso nos permite modelar circuitos como sistemas de equações lineares, onde as incógnitas são as correntes em cada malha.

Considere um circuito com três malhas adjacentes (Malha 1, Malha 2 e Malha 3, da esquerda para a direita), com os seguintes componentes:

* **Malha 1 (Esquerda):** Contém uma fonte de tensão de **5V**, um resistor de **10Ω** (apenas nesta malha) e um resistor de **3Ω** compartilhado com a Malha 2.
* **Malha 2 (Central):** Não possui fonte de tensão. Contém o resistor de **3Ω** (compartilhado com a Malha 1), um resistor de **5Ω** (apenas nesta malha) e um resistor de **2Ω** compartilhado com a Malha 3.
* **Malha 3 (Direita):** Contém uma fonte de tensão de **2V** (com polaridade oposta à da Malha 1), o resistor de **2Ω** (compartilhado com a Malha 2) e um resistor de **8Ω** (apenas nesta malha).

Aplicando a Lei das Malhas para as correntes $I_1$, $I_2$ e $I_3$ em cada malha, obtemos o seguinte sistema:

* **Malha 1:** $5V = (10\Omega)I_1 + (3\Omega)(I_1 - I_2) \implies 13I_1 - 3I_2 = 5$
* **Malha 2:** $0 = (3\Omega)(I_2 - I_1) + (5\Omega)I_2 + (2\Omega)(I_2 - I_3) \implies -3I_1 + 10I_2 - 2I_3 = 0$
* **Malha 3:** $-2V = (2\Omega)(I_3 - I_2) + (8\Omega)I_3 \implies -2I_2 + 10I_3 = -2$

**Sua tarefa:**

1.  Monte o sistema de equações acima na forma matricial $A\mathbf{x} = \mathbf{b}$, onde $\mathbf{x} = [I_1, I_2, I_3]^T$.
2.  Use a função `scipy.linalg.solve` para encontrar o vetor $\mathbf{x}$ de correntes.
3.  Apresente os valores de $I_1$, $I_2$ e $I_3$ em Amperes.

**Exercício 4.** Cadeias de Markov e Estado de Equilíbrio (Autovalores)

Uma [Cadeia de Markov](https://pt.wikipedia.org/wiki/Cadeia_de_Markov) é um modelo matemático que descreve uma sequência de eventos onde a probabilidade de cada evento depende apenas do estado do evento anterior. Podemos usar uma matriz de transição para prever o comportamento de longo prazo de um sistema.

Imagine um mercado com duas empresas, A e B. A cada ano:
* A empresa A retém 90% de seus clientes, e perde 10% para a empresa B.
* A empresa B retém 80% de seus clientes, e perde 20% para a empresa A.

Isso pode ser representado pela matriz de transição $T$:

$$ T = \begin{bmatrix} 0.90 & 0.20 \\ 0.10 & 0.80 \end{bmatrix} $$

Se o estado do mercado (a proporção de clientes em cada empresa) em um ano é o vetor $\mathbf{x}_k = [\%A, \%B]^T$, o estado no ano seguinte será $\mathbf{x}_{k+1} = T \mathbf{x}_k$. O **estado de equilíbrio** do mercado é um vetor $\mathbf{x}_{eq}$ que não muda mais, ou seja, $T \mathbf{x}_{eq} = \mathbf{x}_{eq}$. Isso significa que o estado de equilíbrio é o **autovetor** da matriz de transição associado ao **autovalor $\lambda = 1$**.

**Sua tarefa:**

1. Defina a matriz de transição $T$ usando NumPy.
2. Use `scipy.linalg.eig` para encontrar os autovalores e autovetores de $T$.
3. Identifique qual dos autovetores corresponde ao autovalor $\lambda=1$. (Pode haver pequenas imprecisões numéricas, então procure o autovalor mais próximo de 1).
4. O autovetor encontrado $[v_1, v_2]^T$ representa a razão de clientes no equilíbrio. Para convertê-lo em porcentagens, normalize o vetor para que a soma de seus componentes seja 1 (ou 100%). Calcule $A_{eq} = \frac{v_1}{v_1 + v_2}$ e $B_{eq} = \frac{v_2}{v_1 + v_2}$.
5. Apresente a distribuição de mercado (em porcentagem) que o sistema atingirá no longo prazo.

**Exercício 5.** Interpolação Polinomial (Sistemas Lineares)

A interpolação polinomial consiste em encontrar um polinômio que passe exatamente por um conjunto de pontos. Para $N+1$ pontos, podemos encontrar um único polinômio de grau $N$.

Queremos encontrar o polinômio cúbico $P(x) = c_1 x^3 + c_2 x^2 + c_3 x + c_4$ que passa pelos seguintes quatro pontos: `(-2, -20)`, `(0, 4)`, `(1, 5)` e `(3, 25)`.

Para cada ponto $(x, y)$, podemos escrever uma equação linear substituindo os valores em $P(x) = y$. Por exemplo, para o ponto `(-2, -20)`:
$$c_1(-2)^3 + c_2(-2)^2 + c_3(-2) + c_4 = -20 \implies -8c_1 + 4c_2 - 2c_3 + c_4 = -20$$

**Sua tarefa:**

1. Escreva as quatro equações lineares para os quatro pontos, formando um sistema $A\mathbf{c} = \mathbf{y}$, onde $\mathbf{c} = [c_1, c_2, c_3, c_4]^T$.
2. Crie a matriz $A$ (conhecida como matriz de Vandermonde) e o vetor $\mathbf{y}$ usando NumPy.
3. Use `scipy.linalg.solve` para encontrar os coeficientes do polinômio no vetor $\mathbf{c}$.
4. **Verificação Visual:** Defina uma função Python `P(x)` que implemente o polinômio com os coeficientes encontrados. Em seguida, plote essa função em um gráfico de $x$ entre -3 e 4, e use `plt.scatter` para marcar os quatro pontos originais, confirmando visualmente que seu polinômio passa por eles.
