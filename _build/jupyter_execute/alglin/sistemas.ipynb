## Sistemas Lineares

Um [sistema de equações lineares](https://en.wikipedia.org/wiki/System_of_linear_equations) é um conjunto de equações lineares

\begin{align}
a_{0,0}x_0 + a_{0,1}x_2 + \cdots + a_{0,n}x_n & = b_0 \\\
a_{1,0}x_0 + a_{1,1}x_2 + \cdots + a_{1,n}x_n & = b_1 \\\
& \vdots \\\
a_{m,0}x_0 + a_{m,1}x_2 + \cdots + a_{m,n}x_n & = b_m \\\
\end{align}

Em notação matricial, um sistema linear é $A \mathbf{x}= \mathbf{b}$, onde

$$
A = \begin{bmatrix}
a_{0,0} & a_{0,1} & \cdots & a_{0,n} \\\
a_{1,0} & a_{1,1} & \cdots & a_{1,n} \\\
\vdots & & & \vdots \\\
a_{m,0} & a_{m,1} & \cdots & a_{m,n} \\\
\end{bmatrix}
 \ \ , \ \
\mathbf{x} = \begin{bmatrix}
x_0 \\\ x_1 \\\ \vdots \\\ x_n
\end{bmatrix}
 \ \ , \ \
\mathbf{b} = \begin{bmatrix}
b_0 \\\ b_1 \\\ \vdots \\\ b_m
\end{bmatrix}
$$

## Eliminação Gaussiana

O procedimento geral para resolver um sistema de equações lineares é chamado de [eliminação gaussiana](https://en.wikipedia.org/wiki/Gaussian_elimination). A ideia é realizar operações elementares nas linhas para reduzir o sistema à sua forma escalonada e então resolvê-lo.

### Operações Elementares nas Linhas

As [operações elementares nas linhas](https://en.wikipedia.org/wiki/Elementary_matrix#Elementary_row_operations) incluem:

1. Somar $k$ vezes a linha $j$ à linha $i$.
2. Multiplicar a linha $i$ pelo escalar $k$.
3. Trocar as linhas $i$ e $j$.

Cada uma das operações elementares nas linhas é o resultado da multiplicação por uma matriz elementar (à esquerda).
Para somar $k$ vezes a linha $j$ à linha $i$ em uma matriz $A$, multiplicamos $A$ pela matriz $E$, onde $E$ é igual à matriz identidade, exceto que o elemento $i,j$ é $E_{i,j} = k$. Por exemplo, se $A$ é 3x3 e queremos somar 3 vezes a linha 2 à linha 0 (usando indexação baseada em 0), então

$$
E_1 = \begin{bmatrix}
1 & 0 & 3 \\\
0 & 1 & 0 \\\
0 & 0 & 1
\end{bmatrix}
$$

Vamos verificar o cálculo:

```python
A = np.array([[1,1,2],[-1,3,1],[0,5,2]])
print(A)
```

```python
E1 = np.array([[1,0,3],[0,1,0],[0,0,1]])
print(E1)
```

```python
E1 @ A
```

Para multiplicar a linha $i$ por $k$ em uma matriz $A$, multiplicamos $A$ pela matriz $E$, onde $E$ é igual à matriz identidade, exceto que o elemento $i,i$ é $E_{i,i} = k$. Por exemplo, se $A$ é 3x3 e queremos multiplicar a linha 1 por -2, então

$$
E_2 = \begin{bmatrix}
1 & 0 & 0 \\\
0 & -2 & 0 \\\
0 & 0 & 1
\end{bmatrix}
$$

Vamos verificar o cálculo:

```python
E2 = np.array([[1,0,0],[0,-2,0],[0,0,1]])
print(E2)
```

```python
E2 @ A
```

Finalmente, para trocar a linha $i$ e a linha $j$ em uma matriz $A$, multiplicamos $A$ pela matriz $E$, onde $E$ é igual à matriz identidade, exceto que $E_{i,i} = 0$, $E_{j,j} = 0$, $E_{i,j} = 1$ e $E_{j,i} = 1$. Por exemplo, se $A$ é 3x3 e queremos trocar a linha 1 e a linha 2, então

$$
E_3 = \begin{bmatrix}
1 & 0 & 0 \\\
0 & 0 & 1 \\\
0 & 1 & 0
\end{bmatrix}
$$

Vamos verificar o cálculo:

```python
E3 = np.array([[1,0,0],[0,0,1],[0,1,0]])
print(E3)
```

```python
E3 @ A
```

### Implementação

Vamos escrever funções para implementar as operações elementares nas linhas. Primeiramente, vamos escrever uma função chamada `add_row` que recebe os parâmetros de entrada $A$, $k$, $i$ e $j$ e retorna o array NumPy resultante de somar $k$ vezes a linha $j$ à linha $i$ na matriz $A$. Se $i=j$, vamos dizer que a função escala a linha $i$ por $k+1$, pois este seria o resultado de $k$ vezes a linha $i$ somada à linha $i$.

```python
def add_row(A,k,i,j):
    """Soma k vezes a linha j à linha i na matriz A."""
    n = A.shape[0]
    E = np.eye(n)
    if i == j:
        E[i,i] = k + 1
    else:
        E[i,j] = k
    return E @ A
```

Vamos testar nossa função:

```python
M = np.array([[1,1],[3,2]])
print(M)
```

```python
add_row(M,2,0,1)
```

```python
add_row(M,3,1,1)
```

Vamos escrever uma função chamada `scale_row` que recebe 3 parâmetros de entrada $A$, $k$ e $i$ e retorna a matriz que resulta da multiplicação da linha $i$ por $k$ na matriz $A$.

```python
def scale_row(A,k,i):
    """Multiplica a linha i por k na matriz A."""
    n = A.shape[0]
    E = np.eye(n)
    E[i,i] = k
    return E @ A
```

```python
M = np.array([[3,1],[-2,7]])
print(M)
```

```python
scale_row(M,3,1)
```

```python
A = np.array([[1,1,1],[1,-1,0]])
print(A)
```

```python
scale_row(A,5,1)
```

Vamos escrever uma função chamada `switch_rows` que recebe 3 parâmetros de entrada $A$, $i$ e $j$ e retorna a matriz que resulta da troca das linhas $i$ e $j$ na matriz $A$.

```python
def switch_rows(A,i,j):
    """Troca as linhas i e j na matriz A."""
    n = A.shape[0]
    E = np.eye(n)
    E[i,i] = 0
    E[j,j] = 0
    E[i,j] = 1
    E[j,i] = 1
    return E @ A
```

```python
A = np.array([[1,1,1],[1,-1,0]])
print(A)
```

```python
switch_rows(A,0,1)
```

## Exemplos

### Encontrar a Inversa

Vamos aplicar nossas funções à matriz aumentada $[M \ | \ I]$ para encontrar a inversa da matriz $M$:

```python
M = np.array([[5,4,2],[-1,2,1],[1,1,1]])
print(M)
```

```python
A = np.hstack([M,np.eye(3)])
print(A)
```

```python
A1 = switch_rows(A,0,2)
print(A1)
```

```python
A2 = add_row(A1,1,1,0)
print(A2)
```

```python
A3 = add_row(A2,-5,2,0)
print(A3)
```

```python
A4 = switch_rows(A3,1,2)
print(A4)
```

```python
A5 = scale_row(A4,-1,1)
print(A5)
```

```python
A6 = add_row(A5,-3,2,1)
print(A6)
```

```python
A7 = scale_row(A6,-1/7,2)
print(A7)
```

```python
A8 = add_row(A7,-3,1,2)
print(A8)
```

```python
A9 = add_row(A8,-1,0,2)
print(A9)
```

```python
A10 = add_row(A9,-1,0,1)
print(A10)
```

Vamos verificar se encontramos a inversa $M^{-1}$ corretamente:

```python
Minv = A10[:,3:]
print(Minv)
```

```python
result = Minv @ M
print(result)
```

Sucesso! Podemos ver o resultado mais claramente se arredondarmos para 15 casas decimais:

```python
np.round(result,15)
```

### Resolver um Sistema

Vamos usar nossas funções para realizar a eliminação gaussiana e resolver um sistema de equações lineares $A \mathbf{x} = \mathbf{b}$.

```python
A = np.array([[6,15,1],[8,7,12],[2,7,8]])
print(A)
```

```python
b = np.array([[2],[14],[10]])
print(b)
```

Forme a matriz aumentada $M$:

```python
M = np.hstack([A,b])
print(M)
```

Realize as operações nas linhas:

```python
M1 = scale_row(M,1/6,0)
print(M1)
```

```python
M2 = add_row(M1,-8,1,0)
print(M2)
```

```python
M3 = add_row(M2,-2,2,0)
print(M3)
```

```python
M4 = scale_row(M3,-1/13,1)
print(M4)
```

```python
M5 = add_row(M4,-2,2,1)
print(M5)
```

```python
M6 = scale_row(M5,1/M5[2,2],2)
print(M6)
```

```python
M7 = add_row(M6,-M6[1,2],1,2)
print(M7)
```

```python
M8 = add_row(M7,-M7[0,2],0,2)
print(M8)
```

```python
M9 = add_row(M8,-M8[0,1],0,1)
print(M9)
```

Sucesso! A solução de $Ax=b$ é

```python
x = M9[:,3].reshape(3,1)
print(x)
```

Ou, podemos fazer do jeito fácil...

```python
x = la.solve(A,b)
print(x)
```

## `scipy.linalg.solve`

Estamos interessados principalmente em sistemas lineares $A \mathbf{x} = \mathbf{b}$ onde há uma solução única $\mathbf{x}$. Este é o caso quando $A$ é uma matriz quadrada ($m=n$) e $\mathrm{det}(A) \not= 0$. Para resolver tal sistema, podemos usar a função [`scipy.linalg.solve`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.solve.html).

A função retorna uma solução do sistema de equações $A \mathbf{x} = \mathbf{b}$. Por exemplo:

```python
A = np.array([[1,1],[1,-1]])
print(A)
```

```python
b1 = np.array([2,0])
print(b1)
```

E resolvemos:

```python
x1 = la.solve(A,b1)
print(x1)
```

Note que a saída $\mathbf{x}$ é retornada como um array NumPy 1D quando o vetor $\mathbf{b}$ (o lado direito) é inserido como um array NumPy 1D. Se inserirmos $\mathbf{b}$ como um array NumPy 2D, então a saída é um array NumPy 2D. Por exemplo:

```python
A = np.array([[1,1],[1,-1]])
b2 = np.array([2,0]).reshape(2,1)
x2 = la.solve(A,b2)
print(x2)
```

Finalmente, se o lado direito $\mathbf{b}$ é uma matriz, então a saída é uma matriz do mesmo tamanho. É a solução de $A \mathbf{X} = \mathbf{B}$ quando $\mathbf{B}$ é uma matriz. Por exemplo:

```python
A = np.array([[1,1],[1,-1]])
B = np.array([[2,2],[0,1]])
X = la.solve(A,B)
print(X)
```

```python
Ainv = la.solve(A,np.eye(2))
print(Ainv)
print(Ainv @ A)
```

### Exemplo Simples

Vamos calcular a solução do sistema de equações

\begin{align}
2x + y &= 1 \\\
x + y &= -1
\end{align}

Crie a matriz de coeficientes:

```python
A = np.array([[2,1],[1,1]])
print(A)
```

E o vetor $\mathbf{b}$:

```python
b = np.array([1,-1]).reshape(2,1)
print(b)
```

E resolvemos:

```python
x = la.solve(A,b)
print(x)
```

Podemos verificar a solução calculando a inversa de $A$:

```python
Ainv = la.inv(A)
print(Ainv)
```

E multiplicamos $A^{-1} \mathbf{b}$ para resolver para $\mathbf{x}$:

```python
x = Ainv @ b
print(x)
```

Obtemos o mesmo resultado. Sucesso!

### Inversa ou Resolver

É uma má ideia usar a inversa $A^{-1}$ para resolver $A \mathbf{x} = \mathbf{b}$ se $A$ for grande. É muito caro computacionalmente. Vamos criar uma matriz aleatória grande $A$ e um vetor $\mathbf{b}$ e calcular a solução $\mathbf{x}$ de 2 maneiras:

```python
N = 1000
A = np.random.rand(N,N)
b = np.random.rand(N,1)
```

Verifique os primeiros elementos de $A$:

```python
A[:3,:3]
```

E para $\mathbf{b}$:

```python
b[:4,:]
```

Agora comparamos a velocidade de `scipy.linalg.solve` com `scipy.linalg.inv`:

```python
%%timeit
x = la.solve(A,b)
```

```python
%%timeit
x = la.inv(A) @ b
```

Resolver com `scipy.linalg.solve` é muito mais rápido!
