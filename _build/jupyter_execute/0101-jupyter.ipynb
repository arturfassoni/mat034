# Jupyter Notebooks e Google Colab

## Jupyter Notebook

[Jupyter Notebook](https://jupyter.org) é um tipo de documento web (com extensão .ipynb) para criar e compartilhar documentos que contêm código executável, equações, visualizações e texto explicativo.

## Google Colaboratory

[O Google Colaboratory](https://colab.research.google.com/) é uma plataforma/app para abrir Notebooks Jupyter e executar o código no navegador, sem necessidade de instalar nenhum software.

## Células

Um notebook é uma lista de células. Clique em uma célula para selecioná-la.

Existem dois tipos principais de células: células de código e células de markdown. Pressione `SHIFT+ENTER` para executar o conteúdo de uma célula.

## Adicionando e movendo células
Você pode adicionar novas células usando os botões **+ CÓDIGO** e **+ TEXTO** que aparecem ao passar o mouse entre as células. Esses botões também estão na barra de ferramentas acima do notebook, onde podem ser usados para adicionar uma célula abaixo da célula atualmente selecionada.

Você pode mover uma célula selecionando-a e clicando em **Mover célula para cima** ou **Mover célula para baixo** na barra de ferramentas superior.

Células consecutivas podem ser selecionadas por "seleção de laço", arrastando de fora de uma célula e através do grupo. Células não adjacentes podem ser selecionadas simultaneamente clicando em uma e, em seguida, mantendo pressionada a tecla Ctrl enquanto clica em outra. Da mesma forma, usar Shift em vez de Ctrl selecionará todas as células intermediárias.

## Integração com o Drive

O Colaboratory é integrado ao Google Drive. Ele permite que você compartilhe, comente e colabore no mesmo documento com várias pessoas:

* O botão **COMPARTILHAR** (canto superior direito da barra de ferramentas) permite compartilhar o notebook e controlar as permissões definidas nele.

* **Arquivo->Fazer uma cópia** cria uma cópia do notebook no Drive.

* **Arquivo->Salvar** salva o arquivo no Drive. **Arquivo->Salvar e criar ponto de verificação** fixa a versão para que ela não seja excluída do histórico de revisões.

* **Arquivo->Histórico de revisões** mostra o histórico de revisões do notebook.

## Comentando em uma célula
Você pode comentar em um notebook do Colaboratory como faria em um Documento Google. Os comentários são anexados às células e exibidos ao lado da célula a que se referem. Se você tiver permissões de **apenas comentar**, verá um botão de comentário no canto superior direito da célula ao passar o mouse sobre ela.

Se você tiver permissões de edição ou de comentário, poderá comentar em uma célula de três maneiras:

1. Selecione uma célula e clique no botão de comentário na barra de ferramentas acima do canto superior direito da célula.
2. Clique com o botão direito em uma célula de texto e selecione **Adicionar um comentário** no menu de contexto.
3. Use o atalho **Ctrl+Shift+M** para adicionar um comentário à célula atualmente selecionada.

Você pode resolver e responder a comentários, e pode direcionar comentários a colaboradores específicos digitando *+[endereço de e-mail]* (por exemplo, `+usuario@dominio.com`). Os colaboradores mencionados receberão um e-mail.

O botão Comentário no canto superior direito da página mostra todos os comentários anexados ao notebook.
