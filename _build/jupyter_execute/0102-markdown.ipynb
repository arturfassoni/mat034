# Markdown

[Markdown](https://en.wikipedia.org/wiki/Markdown) é uma linguagem de marcação simples de texto para HTML escrita em texto puro. O Jupyter Notebook reconhece markdown e renderiza o código markdown como HTML. Veja [Markdown](https://daringfireball.net/projects/markdown/) (por John Gruber) e [Ajuda do GitHub Markdown](https://help.github.com/articles/basic-writing-and-formatting-syntax/) para mais informações.

## Texto

| Saída | Sintaxe |
| :---: | :---: |
| *ênfase* | `*ênfase*` |
| **negrito** | `**negrito**` |
| `código` | ``  `código` `` |

## Cabeçalhos

| Saída | Sintaxe |
| :---: | :---: |
| <h1>Título 1</h1> | `# Título 1` |
| <h2>Título 2</h2> | `## Título 2` |
| <h3>Título 3</h3> | `### Título 3` |
| <h4>Título 4</h4> | `#### Título 4` |
| <h5>Título 5</h5> | `##### Título 5` |
| <h6>Título 6</h6> | `###### Título 6` |

## Listas

Crie uma lista ordenada usando números:

```plaintext
1. Teoria dos números
2. Álgebra
3. Equações diferenciais parciais
4. Probabilidade
```

1. Teoria dos números
2. Álgebra
3. Equações diferenciais parciais
4. Probabilidade

Crie uma lista não ordenada usando um asterisco * para cada item:

```plaintext
* Teoria dos números
* Álgebra
* Equações diferenciais parciais
* Probabilidade
```

* Teoria dos números
* Álgebra
* Equações diferenciais parciais
* Probabilidade

Use indentação para criar listas aninhadas:

```plaintext
1. Matemática
  * Cálculo
  * Álgebra Linear
  * Probabilidade
2. Física
  * Mecânica Clássica
  * Relatividade
  * Termodinâmica
3. Biologia
  * Difusão e Osmose
  * Homeostase
  * Imunologia
```

1. Matemática
    * Cálculo
    * Álgebra Linear
    * Probabilidade
2. Física
    * Mecânica Clássica
    * Relatividade
    * Termodinâmica

## Links

Crie um link com a sintaxe `[descrição](url)`. Por exemplo:

```plaintext
[UBC Math](http://www.math.ubc.ca)
```

cria o link [UBC Math](http://www.math.ubc.ca).

## Imagens

Inclua uma imagem usando a sintaxe `![descrição](url)`. Por exemplo:

```plaintext
![Jupyter logo](https://jupyter.org/assets/try/jupyter.png)
```

exibe a imagem

![Jupyter logo](https://jupyter.org/assets/try/jupyter.png)

## Tabelas

Crie uma tabela separando as entradas por caracteres de barra vertical |:

```plaintext
| Operador Python | Descrição  |
| :---: | :---: |
| `+` | adição |
| `-` | subtração |
| `*` | multiplicação |
| `/` | divisão |
| `**` | potência |
```

| Operador Python | Descrição  |
| :---: | :---: |
| `+` | adição |
| `-` | subtração |
| `*` | multiplicação |
| `/` | divisão |
| `**` | potência |

A sintaxe `:---:` especifica o alinhamento (centralizado neste caso) das colunas. Veja mais sobre o [markdown com sabor do GitHub (GitHub flavoured markdown)](https://help.github.com/articles/organizing-information-with-tables/).
