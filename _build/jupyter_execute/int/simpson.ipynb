# Regra de Simpson

## Definição

A [regra de Simpson](https://en.wikipedia.org/wiki/Simpson%27s_rule) usa um polinômio quadrático em cada subintervalo de uma partição para aproximar a função $f(x)$ e calcular a integral definida. Isso é uma melhoria em relação à regra do trapézio, que aproxima $f(x)$ por uma linha reta em cada subintervalo de uma partição.

A fórmula para a regra de Simpson é

$$
S_N(f) = \frac{\Delta x}{3} \sum_{i=1}^{N/2} \left( f(x_{2i-2}) + 4 f(x_{2i-1}) + f(x_{2i}) \right)
$$

onde $N$ é um número *par* de subintervalos de $[a,b]$, $\Delta x = (b - a)/N$ e $x_i = a + i \Delta x$.

## Fórmula de Erro

Vimos que o erro em uma soma de Riemann é inversamente proporcional ao tamanho da partição $N$ e a regra do trapézio é inversamente proporcional a $N^2$. A fórmula de erro no teorema abaixo mostra que a regra de Simpson é ainda melhor, pois o erro é inversamente proporcional a $N^4$.

---

**Teorema** Seja $S_N(f)$ a regra de Simpson

$$
S_N(f) = \frac{\Delta x}{3} \sum_{i=1}^{N/2} \left( f(x_{2i-2}) + 4 f(x_{2i-1}) + f(x_{2i}) \right)
$$

onde $N$ é um número *par* de subintervalos de $[a,b]$, $\Delta x = (b - a)/N$ e $x_i = a + i \Delta x$. O limite do erro é

$$
E_N^S(f) = \left| \ \int_a^b f(x) \, dx - S_N(f) \ \right| \leq \frac{(b-a)^5}{180N^4} K_4
$$

onde $\left| \ f^{(4)}(x) \ \right| \leq K_4$ para todo $x \in [a,b]$.

---

## Implementação

Vamos escrever uma função chamada `simps` que recebe os parâmetros de entrada $f$, $a$, $b$ e $N$ e retorna a aproximação $S_N(f)$. Além disso, vamos atribuir um valor padrão $N=50$.

```python
def simps(f,a,b,N=50):
    '''Approximate the integral of f(x) from a to b by Simpson's rule.

    Simpson's rule approximates the integral \int_a^b f(x) dx by the sum:
    (dx/3) \sum_{k=1}^{N/2} (f(x_{2i-2} + 4f(x_{2i-1}) + f(x_{2i}))
    where x_i = a + i*dx and dx = (b - a)/N.

    Parameters
    ----------
    f : function
        Vectorized function of a single variable
    a , b : numbers
        Interval of integration [a,b]
    N : (even) integer
        Number of subintervals of [a,b]

    Returns
    -------
    float
        Approximation of the integral of f(x) from a to b using
        Simpson's rule with N subintervals of equal length.

    Examples
    --------
    >>> simps(lambda x : 3*x**2,0,1,10)
    1.0
    '''
    if N % 2 == 1:
        raise ValueError("N must be an even integer.")
    dx = (b-a)/N
    x = np.linspace(a,b,N+1)
    y = f(x)
    S = dx/3 * np.sum(y[0:-1:2] + 4*y[1::2] + y[2::2])
    return S
```

Vamos testar nossa função em integrais para as quais conhecemos o valor exato. Por exemplo, sabemos

$$
\int_0^1 3x^2 dx = 1
$$

```python
simps(lambda x : 3*x**2,0,1,10)
```

Teste nossa função novamente com a integral

$$
\int_0^{\pi/2} \sin(x) dx = 1
$$

```python
simps(np.sin,0,np.pi/2,1000)
```

## scipy.integrate.simpson

O subpacote do [SciPy](https://scipy.org/) [scipy.integrate](https://docs.scipy.org/doc/scipy-0.18.1/reference/integrate.html) contém várias funções para aproximar integrais definidas e resolver numericamente equações diferenciais. Vamos importar o subpacote com o nome `spi`.

```python
import scipy.integrate as spi
```

A função `scipy.integrate.simpson` calcula a aproximação de uma integral definida pela regra de Simpson. Consultando a documentação, vemos que tudo o que precisamos fazer é fornecer arrays de valores $x$ e $y$ para o integrando e `scipy.integrate.simpson` retorna a aproximação da integral usando a regra de Simpson.

## Exemplos

### Aproximar ln(2)

Encontre um valor $N$ que garanta que a aproximação pela regra de Simpson $S_N(f)$ da integral

$$
\int_1^2 \frac{1}{x} dx
$$

satisfaça $E_N^S(f) \leq 0.0001$.

Calcule

$$
f^{(4)}(x) = \frac{24}{x^5}
$$

portanto $\left| \, f^{(4)}(x) \, \right| \leq 24$ para todo $x \in [1,2]$ e então

$$
\frac{1}{180N^4} 24 \leq 0.0001 \ \Rightarrow \
\frac{20000}{15N^4} \leq 1 \ \Rightarrow \
\left( \frac{20000}{15} \right)^{1/4} \leq N
$$

Calcule

```python
(20000/15)**0.25
```

Calcule a regra de Simpson com $N=8$ (o menor inteiro par maior que 6.04)

```python
approximation = simps(lambda x : 1/x,1,2,8)
print(approximation)
```

Poderíamos também usar a função `scipy.integrate.simpson` para calcular exatamente o mesmo resultado

```python
N = 8; a = 1; b = 2;
x = np.linspace(a,b,N+1)
y = 1/x
approximation = spi.simpson(y,x)
print(approximation)
```

Verifique se $E_N^S(f) \leq 0.0001$

```python
np.abs(np.log(2) - approximation) <= 0.0001
```
