### **Problema 8: Cálculo do Centro de Massa de uma Placa com Densidade Variável**

O centro de massa $(\bar{x}, \bar{y})$ de uma placa (lâmina) plana que ocupa uma região $D$ com uma função de densidade $\rho(x, y)$ é dado por:

$$\bar{x} = \frac{M_y}{M} \quad , \quad \bar{y} = \frac{M_x}{M}$$

onde:
* Massa total: $M = \iint_D \rho(x, y) \,dA$
* Momento em relação ao eixo y: $M_y = \iint_D x \rho(x, y) \,dA$
* Momento em relação ao eixo x: $M_x = \iint_D y \rho(x, y) \,dA$

Considere uma placa semicircular definida por $x^2 + y^2 \leq 4$ com $y \geq 0$. A densidade da placa não é uniforme e varia com a distância ao eixo y, sendo dada por $\rho(x, y) = 1 + 0.5x^2$.

**Tarefas:**

1.  **Configuração das Integrais:** Converta as integrais duplas acima em integrais iteradas. Por simetria, você pode deduzir que $\bar{x} = 0$. Você precisará configurar as integrais para a massa $M$ e o momento $M_x$:

    $$M = \int_{-2}^{2} \int_{0}^{\sqrt{4-x^2}} (1 + 0.5x^2) \,dy\,dx$$

    $$M_x = \int_{-2}^{2} \int_{0}^{\sqrt{4-x^2}} y(1 + 0.5x^2) \,dy\,dx$$

2.  **Simplificação:** Resolva as integrais internas (em relação a $y$) **analiticamente**. Isso resultará em duas integrais simples (em relação a $x$).

3.  **Cálculo Numérico:** Use a **regra de Simpson** (`simps` ou `scipy.integrate.simps`) com `N=200` para calcular os valores numéricos das duas integrais resultantes da etapa 2.

4.  **Resultado Final:** Calcule e apresente a coordenada $\bar{y}$ do centro de massa.

### **Problema 9: O Problema da Braquistócrona e o Tempo de Queda**

A [braquistócrona](https://pt.wikipedia.org/wiki/Braquist%C3%B3crona) é a curva entre dois pontos que um objeto em queda livre (sem atrito, sob gravidade constante) percorre no menor tempo possível. Essa curva não é uma reta, mas sim uma cicloide.

As equações paramétricas de uma cicloide (com $r=1$) que passa pela origem são:

$$x(\theta) = r(\theta - \sin\theta), \quad y(\theta) = r(1 - \cos\theta). $$

O tempo de descida ao longo de uma curva $(x(\theta), y(\theta))$ de $\theta_A$ até $\theta_B$ é:

$$T = \int_{\theta_A}^{\theta_B} \frac{\sqrt{x'(\theta)^2 + y'(\theta)^2}}{\sqrt{2gy(\theta)}} d\theta$$

**Tarefas:**

1.  **Cálculo das Derivadas:** Calcule as derivadas $x'(\theta)$ e $y'(\theta)$ **analiticamente**.

2.  **Função do Integrando:** Calcule a função do integrando $f(\theta) = \frac{\sqrt{x'(\theta)^2 + y'(\theta)^2}}{\sqrt{2gy(\theta)}}$ analiticamente e então a defina como uma função em Python. Assuma $r=1$ e $g=9.81$.

3.  **Tempo na Cicloide:** Calcule numericamente o tempo de descida ao longo da cicloide desde o ponto A (origem, $\theta_A = 0$) até o ponto B ($x \approx 5.14$, $y \approx 1.27$, que corresponde a $\theta_B = 2.5$ radianos). Use uma **técnica de integração numérica** de sua escolha (por exemplo, `scipy.integrate.quad` ou sua função `simps` com N grande).

4.  **Comparação:** Calcule o tempo de descida para um objeto que cai ao longo de uma **rampa reta** entre os mesmos pontos A e B. A fórmula para o tempo em uma rampa reta de comprimento $L$ e inclinação $\alpha$ é $T_{reta} = \sqrt{2L / (g \sin\alpha)}$.

5.  **Conclusão (em uma célula Markdown):** Compare os dois tempos. Seu resultado confirma que a cicloide é a trajetória mais rápida?

### **Problema 10: Análise do Potencial de Lennard-Jones**

O [potencial de Lennard-Jones](https://pt.wikipedia.org/wiki/Potencial_de_Lennard-Jones) é um modelo matemático que descreve a energia potencial $V$ da interação entre duas partículas neutras em função da distância $r$ entre elas:

$$V(r) = 4\epsilon \left[ \left(\frac{\sigma}{r}\right)^{12} - \left(\frac{\sigma}{r}\right)^{6} \right]$$

A força entre as partículas é $F(r) = -V'(r)$. Uma posição de equilíbrio estável ocorre no ponto $r_0$ onde a energia potencial é mínima, ou seja, onde a força é nula, $F(r_0) = -V'(r_0) = 0$.

Para esta questão, use os parâmetros para o Argônio: $\sigma = 3.4 \times 10^{-10}$ m e $\epsilon = 1.65 \times 10^{-21}$ J.

**Tarefas:**

1.  **Força e Equilíbrio:**

    a) Calcule a primeira derivada $V'(r)$ **analiticamente** para obter a fórmula da força $F(r)$.

    b) Use um **método de busca de raiz** de sua escolha para encontrar a distância de equilíbrio $r_0$ onde $F(r_0)=0$. Use uma tolerância de $10^{-12}$.

2.  **Visualização:** Plote o potencial $V(r)$ em função de $r$ (para $r$ de $3 \times 10^{-10}$ a $8 \times 10^{-10}$ m). Marque no gráfico o ponto de mínimo $(r_0, V(r_0))$ que você encontrou.

3.  **Frequência de Oscilação:** Para pequenas oscilações em torno do ponto de equilíbrio, o sistema se comporta como um oscilador harmônico com uma "constante de mola" $k = V''(r_0)$.

    a) Use **diferenciação numérica** para calcular a segunda derivada $V''(x)$ no ponto de equilíbrio $r_0$ para encontrar a constante $k$.

    b) A frequência angular de oscilação é $\omega = \sqrt{k/\mu}$, onde $\mu$ é a massa reduzida. Para duas partículas de Argônio, $\mu \approx 3.32 \times 10^{-26}$ kg. Calcule a frequência angular $\omega$.

4.  **Relatório Final:** Apresente a distância de equilíbrio $r_0$, a energia potencial mínima $V(r_0)$, a constante de mola efetiva $k$ e a frequência de oscilação $\omega$.
