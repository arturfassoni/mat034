# LaTeX

[LaTeX](https://www.latex-project.org/) é um ambiente de programação para a produção de documentos científicos. O Jupyter Notebook reconhece o código LaTeX escrito em células de markdown e renderiza os símbolos matemáticos no navegador usando a biblioteca JavaScript [MathJax](https://www.mathjax.org/).

## Matemática Em Linha e em Destaque

Envolva o código LaTeX com cifrões `$ ... $` para exibir matemática em linha. Por exemplo, o código `$\int_a^b f(x) = F(b) - F(a)$` é renderizado em linha como $ \int_a^b f(x) dx = F(b) - F(a) $.

Envolva o código LaTeX com cifrões duplos `$$ ... $$` para exibir expressões em um parágrafo centralizado. Por exemplo:

```LaTeX
$$f'(a) = \lim_{x \to a} \frac{f(x) - f(a)}{x-a}$$
```

é renderizado como

$$f'(a) = \lim_{x \to a} \frac{f(x) - f(a)}{x-a}$$

Consulte o [LaTeX WikiBook](https://en.wikibooks.org/wiki/LaTeX) para mais informações (especialmente a seção sobre [matemática](https://en.wikibooks.org/wiki/LaTeX/Mathematics)).

## Matrizes e Parênteses

Crie uma matriz sem parênteses:

```LaTeX
$$\begin{matrix} a & b \\ c & d \end{matrix}$$
```

$$
\begin{matrix} a & b \\\ c & d \end{matrix}
$$

Crie uma matriz com parênteses redondos:

```LaTeX
$$\begin{pmatrix} a & b \\ c & d \end{pmatrix}$$
```

$$
\begin{pmatrix} a & b \\\ c & d \end{pmatrix}
$$

Crie uma matriz com colchetes:

```LaTeX
$$\begin{bmatrix} 1 & 2 & 1 \\ 3 & 0 & 1 \\ 0 & 2 & 4 \end{bmatrix}$$
```

$$
\begin{bmatrix}
1 & 2 & 1 \\\
3 & 0 & 1 \\\
0 & 2 & 4
\end{bmatrix}
$$

Use `\left` e `\right` para envolver qualquer expressão em parênteses/colchetes:

```LaTeX
$$\left( \frac{p}{q} \right)$$
```

$$\left( \frac{p}{q} \right)$$

## Exemplos

### Derivada

A [derivada](https://en.wikipedia.org/wiki/Derivative) $f'(a)$ da função $f(x)$ no ponto $x=a$ é o limite:

```LaTeX
$$f'(a) = \lim_{x \to a} \frac{f(x) - f(a)}{x - a}$$
```

$$f'(a) = \lim_{x \to a} \frac{f(x) - f(a)}{x - a}$$

### Continuidade

Uma função $f(x)$ é [contínua](https://en.wikipedia.org/wiki/Continuous_function) em um ponto $x=a$ se:

```LaTeX
$$\lim_{x \to a^-} f(x) = f(a) = \lim_{x \to a^+} f(x)$$
```

$$\lim_{x \to a^-} f(x) = f(a) = \lim_{x \to a^+} f(x)$$

### Série de MacLaurin

A [série de MacLaurin](https://en.wikipedia.org/wiki/Taylor_series) para $e^x$ é:

```LaTeX
$$e^x = \sum_{k=0}^{\infty} \frac{x^k}{k!}$$
```

$$e^x = \sum_{k=0}^{\infty} \frac{x^k}{k!}$$

### Matriz Jacobiana

A [matriz Jacobiana](https://en.wikipedia.org/wiki/Jacobian_matrix_and_determinant) da função $\mathbf{f}(x_1, \dots, x_n)$ é:

```LaTeX
$$
\mathbf{J}
=
\frac{d \mathbf{f}}{d \mathbf{x}}
=
\left[ \frac{\partial \mathbf{f}}{\partial x_1}
\cdots \frac{\partial \mathbf{f}}{\partial x_n} \right]
=
\begin{bmatrix}
\frac{\partial f_1}{\partial x_1} & \cdots &
\frac{\partial f_1}{\partial x_n} \\
\vdots & \ddots & \vdots \\
\frac{\partial f_m}{\partial x_1} & \cdots &
\frac{\partial f_m}{\partial x_n}
\end{bmatrix}
$$
```

$$
\mathbf{J} = \frac{d \mathbf{f}}{d \mathbf{x}} =
\left[ \frac{\partial \mathbf{f}}{\partial x_1}
\cdots \frac{\partial \mathbf{f}}{\partial x_n} \right] =
\begin{bmatrix}
\frac{\partial f_1}{\partial x_1} & \cdots & \frac{\partial f_1}{\partial x_n} \\\
\vdots & \ddots & \vdots \\\
\frac{\partial f_m}{\partial x_1} & \cdots & \frac{\partial f_m}{\partial x_n}
\end{bmatrix}
$$
