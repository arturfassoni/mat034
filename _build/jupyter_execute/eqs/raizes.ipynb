# Encontrando raízes

[Encontrar raízes](https://en.wikipedia.org/wiki/Root-finding_algorithm) refere-se ao problema geral de procurar uma solução para uma equação $F(x)=0$ para alguma função $F(x)$. Este é um problema muito geral e que aparece com frequência na matemática! Por exemplo, se quisermos otimizar uma função $f(x)$, precisamos encontrar os pontos críticos e, portanto, resolver a equação $f'(x)=0$.

Existem poucos exemplos onde há métodos exatos para encontrar soluções. Por exemplo, a [fórmula de Bhaskara](https://en.wikipedia.org/wiki/Quadratic_formula)

$$
x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
$$

nos dá um método exato para encontrar as raízes da equação

$$
ax^2 + bx + c = 0
$$

Existe uma [fórmula geral para resolver uma equação cúbica](https://en.wikipedia.org/wiki/Cubic_function#General_solution_to_the_cubic_equation_with_real_coefficients) e até mesmo uma [equação quártica (grau 4)](https://en.wikipedia.org/wiki/Quartic_function) (mas a fórmula é complicada demais para ser útil).

No entanto, não existe uma fórmula para um [polinômio quíntico (grau 5)](https://en.wikipedia.org/wiki/Quintic_function). E há *muitos* outros exemplos de equações sem nenhum método conhecido para resolvê-las exatamente.

O que podemos fazer? Usar métodos numéricos para encontrar soluções aproximadas.
