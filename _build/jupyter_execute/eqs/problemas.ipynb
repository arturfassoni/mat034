**Exercício A (Ponto de Mínima Distância)**

Este é um problema clássico de otimização. Dado um ponto $P = (2, -1)$ e a parábola $f(x) = x^2 - 4x + 5$, queremos encontrar o ponto $(x, f(x))$ na parábola que está mais próximo de $P$.

1.  A distância entre um ponto $(x, f(x))$ na curva e o ponto $P$ é dada por $d(x) = \sqrt{(x - 2)^2 + (f(x) - (-1))^2}$. Para simplificar, minimizamos o quadrado da distância: $D(x) = d(x)^2 = (x - 2)^2 + (x^2 - 4x + 6)^2$.
2.  Para encontrar o mínimo de $D(x)$, precisamos encontrar a raiz de sua derivada, $D'(x) = 0$.
3.  **Sua tarefa:**
    a) Calcule a expressão analítica para $D'(x)$.
    b) Plote o gráfico de $D'(x)$ para estimar visualmente onde a raiz se encontra.
    c) Use um dos métodos de busca de raiz (bissecção, secante ou Newton) para encontrar o valor de $x$ que zera $D'(x)$ com uma tolerância de $10^{-7}$.
    d) Reporte as coordenadas $(x, y)$ do ponto na parábola mais próximo de $P$.

**Exercício B (Lançamento de Projétil)**

Um projétil é lançado de uma altura de 10 metros com uma velocidade inicial de 50 m/s em um ângulo $\theta$. A altura do projétil em função da distância horizontal $x$ é dada por:

$$y(x) = 10 + (\tan\theta) x - \frac{g}{2 v_0^2 \cos^2\theta} x^2$$

onde $g = 9.8 \, m/s^2$ e $v_0 = 50 \, m/s$.

O objetivo é encontrar o ângulo de lançamento $\theta$ (em radianos) para que o projétil atinja um alvo a uma distância de **200 metros** (ou seja, $y(200) = 0$).

1.  **Sua tarefa:**
    a) Defina uma função $f(\theta) = 10 + (\tan\theta) \cdot 200 - \frac{g}{2 \cdot 50^2 \cos^2\theta} \cdot 200^2$. Você precisa encontrar a raiz desta função, $f(\theta) = 0$.
    b) Plote o gráfico de $f(\theta)$ para $\theta$ entre $0$ e $\pi/2$ para visualizar as possíveis soluções (ângulos). Você verá que existem duas!
    c) Use um método de busca de raiz (como o da secante) para encontrar os dois ângulos $\theta_1$ e $\theta_2$ com uma precisão de $10^{-6}$. Escolha intervalos iniciais apropriados com base no seu gráfico.
    d) Para cada um dos dois ângulos encontrados, encontre a inclinação da trajetória, $y'(x)$, no momento do impacto (em $x=200$). Qual dos dois ângulos resulta em um impacto mais "rasante" (ou seja, com $y'(200)$ mais próximo de zero)?
