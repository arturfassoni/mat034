**Exercício A (Comparando Taxas de Convergência)**

O objetivo deste projeto é visualizar a superioridade da Regra de Simpson em relação à Regra do Trapézio e à Soma de Riemann (ponto médio).

1.  Considere a integral $\int_{0}^{\pi} \sin(x) dx$, cujo valor exato é 2.
2.  Crie uma lista de valores para `N` (número de subintervalos), por exemplo: `N_valores = [10, 20, 50, 100, 200, 500, 1000]`.
3.  Para cada valor de `N` na lista, calcule a integral usando:
    a) A soma de Riemann pelo ponto médio.
    b) A regra do trapézio.
    c) A regra de Simpson.
4.  Para cada cálculo, armazene o **erro absoluto** em relação ao valor exato (2) em listas separadas (uma para cada método).
5.  Crie um único gráfico **log-log** (`plt.loglog()`) que plote o erro de cada método em função de `N`.
6.  Adicione uma legenda para identificar as curvas de erro de cada método. O que o gráfico lhe diz sobre a rapidez com que o erro de cada método diminui à medida que `N` aumenta? (A inclinação das retas no gráfico log-log está relacionada à ordem do erro).

**Exercício B (Período de um Pêndulo Não Linear)**

Para pequenas oscilações, o período de um pêndulo simples é $T \approx 2\pi\sqrt{L/g}$, independente da amplitude. No entanto, para oscilações maiores, o período real `T` depende da amplitude máxima $\theta_{max}$ (em radianos) e é dado pela integral elíptica:

$$T = 4 \sqrt{\frac{L}{g}} \int_0^{\pi/2} \frac{d\phi}{\sqrt{1 - \sin^2(\frac{\theta_{max}}{2}) \sin^2(\phi)}}$$

Sua tarefa é investigar como o período de um pêndulo de comprimento $L=1$ metro varia com a amplitude. Assuma $g = 9.81 \, m/s^2$.

1.  Escreva uma função `periodo_pendulo(theta_max, L=1, N=100)` que calcule a integral acima usando a **regra de Simpson**.
    * A função do integrando será $f(\phi) = \frac{1}{\sqrt{1 - k \sin^2(\phi)}}$, onde $k = \sin^2(\frac{\theta_{max}}{2})$.
    * A função deve retornar o período `T` completo.
2.  Crie um array de ângulos `theta_valores` de $0.1$ a $3.1$ radianos (quase 180 graus) usando `np.linspace()`.
3.  Use um loop `for` para calcular o período `T` para cada `theta_max` em `theta_valores` e armazene os resultados.
4.  Plote o Período `T` (eixo y) em função da Amplitude Máxima `theta_max` (em graus, para melhor intuição) no eixo x.
5.  No mesmo gráfico, plote uma linha horizontal mostrando o período para pequenas oscilações, $T_{pequeno} = 2\pi\sqrt{L/g}$.
6.  Adicione um título e rótulos aos eixos. O que você pode concluir sobre como o período de um pêndulo real muda com a amplitude?
