# Variáveis

Assim como as variáveis familiares $x$ e $y$ na matemática, usamos variáveis na programação para manipular valores facilmente. O operador de atribuição `=` atribui valores a variáveis em Python. Escolha nomes de variáveis descritivos e evite palavras reservadas especiais do Python.

## Atribuir Valores a Variáveis

Atribua um valor a uma variável usando o operador de atribuição `=`. Por exemplo, atribua o inteiro 2 à variável `x`:

```python
x = 2
```

O operador de atribuição não produz nenhuma saída e, portanto, a célula acima não gera nenhuma saída em um notebook Jupyter. Use a função embutida `print` para exibir o valor atribuído a uma variável:

```python
print(x)
```

Calcule novos valores usando variáveis e operadores:

```python
1 + x + x**2 + x**3
```

Use a função embutida `type` para verificar o tipo de dado do valor atribuído a uma variável:

```python
pi = 3.14159
type(pi)
```

## Convenções de Nomenclatura

Podemos usar qualquer conjunto de letras, números e underscores para criar nomes de variáveis, no entanto, um nome de variável *não pode* começar com um número. Existem muitos tipos diferentes de convenções de nomenclatura e nos referimos ao [Guia de Estilo para Código Python (PEP8)](https://www.python.org/dev/peps/pep-0008/#naming-conventions) para um resumo.

Neste livro, usamos nomes de variáveis no formato `minusculas_com_underscores` e nomes de variáveis de uma única letra minúscula, como `x`. É uma boa prática usar nomes de variáveis descritivos para tornar seu código mais legível para outras pessoas.

Por exemplo, a distância de Vancouver a Halifax ao longo da Rodovia Trans-Canadá é de aproximadamente 5799 quilômetros. Escrevemos o seguinte código para converter este valor em milhas:

```python
distance_km = 5799
miles_per_km = 0.6214
distance_miles = distance_km * miles_per_km
print(distance_miles)
```

## Nomes a Evitar

É uma boa prática usar nomes de variáveis que descrevam o valor atribuído a elas. No entanto, existem palavras que não devemos usar como nomes de variáveis porque essas palavras já têm um significado especial em Python.

### Palavras Reservadas

Abaixo estão resumidas as [palavras reservadas no Python 3](https://docs.python.org/3.3/reference/lexical_analysis.html#keywords). O Python gerará um erro se você tentar atribuir um valor a qualquer uma dessas palavras-chave e, portanto, você deve evitá-las como nomes de variáveis.

| | | | | |
| :---: | :---: | :---: | :---: | :---: |
| `False` | `class` | `finally` | `is` | `return` |
| `None` |  `continue` | `for` | `lambda` | `try` |
| `True` | `def` | `from` |  `nonlocal` | `while` |
| `and` | `del` | `global` | `not` | `with` |
| `as` | `elif` | `if` | `or` | `yield` |
| `assert` | `else` | `import` | `pass` | `break` |
| `except` | `in` | `raise` | | |

### Nomes de Funções Embutidas

Existem várias funções que estão incluídas na biblioteca padrão do Python. *Não* use os nomes dessas funções como nomes de variáveis, caso contrário, a referência à função embutida será perdida. Por exemplo, não use `sum`, `min`, `max`, `list` ou `sorted` como nome de variável. Veja a lista completa de [funções embutidas](https://docs.python.org/3/library/functions.html).
