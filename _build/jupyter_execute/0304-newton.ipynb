# Método de Newton

O [método de Newton](https://en.wikipedia.org/wiki/Newton%27s_method) é um método para encontrar raízes que usa aproximação linear. Em particular, supomos uma solução $x_0$ da equação $f(x)=0$, calculamos a aproximação linear de $f(x)$ em $x_0$ e, em seguida, encontramos a interceptação com o eixo $x$ da aproximação linear.

## Fórmula

Seja $f(x)$ uma função diferenciável. Se $x_0$ está próximo de uma solução de $f(x)=0$, então podemos aproximar $f(x)$ pela reta tangente em $x_0$ e calcular a interceptação com o eixo $x$ da reta tangente. A equação da reta tangente em $x_0$ é

$$
y = f'(x_0)(x - x_0) + f(x_0)
$$

A interceptação com o eixo $x$ é a solução $x_1$ da equação

$$
0 = f'(x_0)(x_1 - x_0) + f(x_0)
$$

e resolvemos para $x_1$

$$
x_1 = x_0 - \frac{f(x_0)}{f'(x_0)}
$$

Se implementarmos este procedimento repetidamente, obtemos uma sequência dada pela fórmula recursiva

$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

que (potencialmente) converge para uma solução da equação $f(x)=0$.

## Vantagens/Desvantagens

Quando converge, o método de Newton geralmente converge muito rapidamente, e essa é sua principal vantagem. No entanto, não há garantia de que o método de Newton irá convergir, e isso é obviamente uma grande desvantagem, especialmente em comparação com os métodos da bissecção e da secante, que têm garantia de convergir para uma solução (desde que comecem com um intervalo contendo uma raiz).

O método de Newton também requer o cálculo de valores da derivada da função em questão. Isso é potencialmente uma desvantagem se a derivada for difícil de calcular.

Os critérios de parada para o método de Newton diferem dos métodos da bissecção e da secante. Nesses métodos, sabemos o quão perto estamos de uma solução porque estamos calculando intervalos que contêm uma solução. No método de Newton, não sabemos o quão perto estamos de uma solução. Tudo o que podemos calcular é o valor $f(x)$ e, portanto, implementamos um critério de parada com base em $f(x)$.

Finalmente, não há garantia de que o método convirja para uma solução e devemos definir um número máximo de iterações para que nossa implementação termine se não encontrarmos uma solução.

## Implementação

Vamos escrever uma função chamada `newton` que recebe 5 parâmetros de entrada `f`, `Df`, `x0`, `epsilon` e `max_iter` e retorna uma aproximação de uma solução de $f(x)=0$ pelo método de Newton. A função pode terminar de 3 maneiras:

1. Se `abs(f(xn)) < epsilon`, o algoritmo encontrou uma solução aproximada e retorna `xn`.
2. Se `f'(xn) == 0`, o algoritmo para e retorna `None`.
3. Se o número de iterações exceder `max_iter`, o algoritmo para e retorna `None`.

```python
def newton(f,Df,x0,epsilon,max_iter):
    '''Approximate solution of f(x)=0 by Newton's method.

    Parameters
    ----------
    f : function
        Function for which we are searching for a solution f(x)=0.
    Df : function
        Derivative of f(x).
    x0 : number
        Initial guess for a solution f(x)=0.
    epsilon : number
        Stopping criteria is abs(f(x)) < epsilon.
    max_iter : integer
        Maximum number of iterations of Newton's method.

    Returns
    -------
    xn : number
        Implement Newton's method: compute the linear approximation
        of f(x) at xn and find x intercept by the formula
            x = xn - f(xn)/Df(xn)
        Continue until abs(f(xn)) < epsilon and return xn.
        If Df(xn) == 0, return None. If the number of iterations
        exceeds max_iter, then return None.

    Examples
    --------
    >>> f = lambda x: x**2 - x - 1
    >>> Df = lambda x: 2*x - 1
    >>> newton(f,Df,1,1e-8,10)
    Found solution after 5 iterations.
    1.618033988749989
    '''
    xn = x0
    for n in range(0,max_iter):
        fxn = f(xn)
        if abs(fxn) < epsilon:
            print('Found solution after',n,'iterations.')
            return xn
        Dfxn = Df(xn)
        if Dfxn == 0:
            print('Zero derivative. No solution found.')
            return None
        xn = xn - fxn/Dfxn
    print('Exceeded maximum iterations. No solution found.')
    return None
```

### Razão Superáurea

Vamos testar nossa função `newton` no polinômio $p(x) = x^3 - x^2 - 1$ para aproximar a [razão superáurea](https://en.wikipedia.org/wiki/Supergolden_ratio).

```python
p = lambda x: x**3 - x**2 - 1
Dp = lambda x: 3*x**2 - 2*x
approx = newton(p,Dp,1,1e-10,10)
print(approx)
```
