# Diferenciação Numérica

## Derivada

A [derivada](https://en.wikipedia.org/wiki/Derivative) de uma função $f(x)$ em $x=a$ é o limite

$$
f'(a) = \lim_{h \to 0} \frac{f(a+h) - f(a)}{h}
$$

## Fórmulas de Diferença

Existem 3 principais [fórmulas de diferença](https://en.wikipedia.org/wiki/Finite_difference#Forward,_backward,_and_central_differences) para aproximar numericamente as derivadas.

A *fórmula de diferença progressiva* com passo $h$ é

$$
f'(a) \approx \frac{f(a + h) - f(a)}{h}
$$

A *fórmula de diferença regressiva* com passo $h$ é

$$
f'(a) \approx \frac{f(a) - f(a - h)}{h}
$$

A *fórmula de diferença central* com passo $h$ é a média das fórmulas de diferença progressiva e regressiva

$$
f'(a) \approx \frac{1}{2} \left( \frac{f(a + h) - f(a)}{h} + \frac{f(a) - f(a - h)}{h} \right) = \frac{f(a + h) - f(a - h)}{2h}
$$

## Implementação

Vamos escrever uma função chamada `derivative` que recebe os parâmetros de entrada `f`, `a`, `method` e `h` (com valores padrão `method='central'` e `h=0.01`) e retorna a fórmula de diferença correspondente para $f'(a)$ com passo $h$.

```python
def derivative(f,a,method='central',h=0.01):
    '''Compute the difference formula for f'(a) with step size h.

    Parameters
    ----------
    f : function
        Vectorized function of one variable
    a : number
        Compute derivative at x = a
    method : string
        Difference formula: 'forward', 'backward' or 'central'
    h : number
        Step size in difference formula

    Returns
    -------
    float
        Difference formula:
            central: f(a+h) - f(a-h))/2h
            forward: f(a+h) - f(a))/h
            backward: f(a) - f(a-h))/h
    '''
    if method == 'central':
        return (f(a + h) - f(a - h))/(2*h)
    elif method == 'forward':
        return (f(a + h) - f(a))/h
    elif method == 'backward':
        return (f(a) - f(a - h))/h
    else:
        raise ValueError("Method must be 'central', 'forward' or 'backward'.")
```

Vamos testar nossa função em algumas funções simples. Por exemplo, sabemos que

$$
\left. \frac{d}{dx} \left( \cos x \right) \, \right|_{x=0} = -\sin(0) = 0
$$

e calculamos

```python
derivative(np.cos,0)
```

```python
derivative(np.cos,0,method='forward',h=1e-2)
```

```python
derivative(np.cos,0,method='backward',h=1e-2)
```

Também sabemos que

$$
\left. \frac{d}{dx} \left( e^x \right) \, \right|_{x=0} = e^0 = 1
$$

e calculamos

```python
derivative(np.exp,0,h=0.0001)
```

```python
derivative(np.exp,0,method='backward',h=0.0001)
```

Note que nossa função pode receber um array de entradas para `a` e retornar as derivadas para cada valor de `a`. Por exemplo, podemos plotar a derivada de $\sin(x)$:

```python
x = np.linspace(0,5*np.pi,100)
dydx = derivative(np.sin,x)

dYdx = np.cos(x)

plt.figure(figsize=(12,5))
plt.plot(x,dydx,'r.',label='Central Difference')
plt.plot(x,dYdx,'b',label='True Value')

plt.title('Central Difference Derivative of y = cos(x)')
plt.legend(loc='best')
plt.show()
```

Vamos calcular e plotar a derivada de uma função complicada

$$
y=\left(\frac{4x^2+2x+1}{x+2e^x}\right)^x
$$

```python
x = np.linspace(0,6,100)
f = lambda x: ((4*x**2 + 2*x + 1)/(x + 2*np.exp(x)))**x
y = f(x)
dydx = derivative(f,x)

plt.figure(figsize=(12,5))
plt.plot(x,y,label='y=f(x)')
plt.plot(x,dydx,label="Central Difference y=f'(x)")
plt.legend()
plt.grid(True)

plt.show()
```

## Fórmulas de Erro
    
Surgem perguntas naturais: quão boas são as aproximações dadas pelas fórmulas de diferença progressiva, regressiva e central? Derivamos as fórmulas de erro a partir do [Teorema de Taylor](https://en.wikipedia.org/wiki/Taylor%27s_theorem).

---

**Teorema.** O polinômio de Taylor de grau $n$ de $f(x)$ em $x=a$ com termo de resto é

$$
f(x) = f(a) + f'(a)(x - a) + \frac{f''(a)}{2}(x-a)^2 + \cdots + \frac{f^{(n)}(a)}{n!}(x-a)^n + \frac{f^{(n+1)}(c)}{(n+1)!}(x-a)^{n+1}
$$

para algum valor $c$ entre $x$ e $a$.

---

**Teorema.** O erro da fórmula de diferença progressiva é

$$
\left| \, \frac{f(a+h) - f(a)}{h} - f'(a) \, \right| \leq \frac{hK_2}{2}
$$

onde $\left| \, f''(x) \, \right| \leq K_2$ para todo $x \in [a,a+h]$. A mesma fórmula de erro é válida para a fórmula de diferença regressiva.

*Prova*. Observe a fórmula de Taylor de grau 1:

$$
f(x) = f(a) + f'(a)(x-a) + \frac{f''(c)}{2}(x-a)^{2}
$$

Seja $x = a+h$ e manipule a fórmula

\begin{align}
f(a+h) &= f(a) + f'(a)h + \frac{f''(c)}{2}h^{2} \\\
f(a+h) - f(a) &= f'(a)h + \frac{f''(c)}{2}h^{2} \\\
\frac{f(a+h) - f(a)}{h} &= f'(a) + \frac{f''(c)}{2}h \\\
\frac{f(a+h) - f(a)}{h} - f'(a) &= \frac{f''(c)}{2}h
\end{align}

Seja $K_2$ tal que $\left| \, f''(x) \, \right| \leq K_2$ para todo $x \in [a,a+h]$ e vemos o resultado.

---

**Teorema.** O erro da fórmula de diferença central é:

$$
\left| \frac{f(a+h) - f(a-h)}{2h} - f'(a) \right| \leq \frac{h^2K_3}{6}
$$

onde $|f'''(x)| \leq K_3$ para todo $x \in [a-h,a+h]$.

*Prova*. Observe o polinômio de Taylor de grau 2:

$$
f(x) = f(a) + f'(a)(x-a) + \frac{f''(a)}{2}(x-a)^2 + \frac{f'''(c)}{6}(x-a)^{3}
$$

Seja $x = a + h$ e também $x = a - h$ e escreva:

\begin{align}
f(a+h) &= f(a) + f'(a)h + \frac{f''(a)}{2}h^2 + \frac{f'''(c_1)}{6}h^{3} \\\
f(a-h) &= f(a) - f'(a)h + \frac{f''(a)}{2}h^2 - \frac{f'''(c_2)}{6}h^{3} \\\
f(a+h) - f(a-h) &= 2 f'(a)h + \frac{f'''(c_1)}{6}h^{3} + \frac{f'''(c_2)}{6}h^{3} \\\
\frac{f(a+h) - f(a-h)}{2h} - f'(a) &= \frac{f'''(c_1) + f'''(c_2)}{12}h^{2}
\end{align}

Note que $f'''(x)$ é contínua (por hipótese) e $(f'''(c_1) + f'''(c_2))/2$ está entre $f'''(c_1)$ e $f'''(c_2)$ e, portanto, existe algum $c$ entre $c_1$ e $c_2$ tal que

$$
f'''(c) = \frac{f'''(c_1) + f'''(c_2)}{2}
$$

pelo Teorema do Valor Intermediário. Seja $K_3$ tal que $\left| \, f'''(x) \, \right| \leq K_3$ para todo $x \in [a-h,a+h]$ e vemos o resultado.

## scipy.differentiate.derivative

A função do [SciPy](https://scipy.org/) `scipy.differentiate.derivative` calcula derivadas usando a fórmula de diferença central.

```python
from scipy.differentiate import derivative
```

```python
f = np.exp
df = np.exp  # true derivative
x = np.arange(0, 4)
derivative(np.exp,x)
```

```python
res = derivative(f, x)
print(res.df)  # approximation of the derivative
print(res.error)  # estimate of the error
print(abs(res.df - df(x)))  # true error
```

## Derivadas de ordem superior

Derivadas de ordem superior em Python podem ser calculadas usando vários métodos, dependendo se é necessária uma diferenciação simbólica ou numérica.

**Derivadas numéricas de ordem superior:**

Para aproximação numérica de derivadas de ordem superior de dados discretos ou funções onde uma forma analítica não está disponível, `numpy.gradient()` pode ser usado iterativamente.

Note que a função `numpy.gradient()` calcula a derivada numerica baseada no grid de pontos x,y fornecido:



```python
import numpy as np

# Example function: y = x^3
x = np.linspace(0, 10, 200+1)
y = x**3

# First derivative
dydx = np.gradient(y, x)

# Second derivative
d2ydx2 = np.gradient(dydx, x)

# Third derivative
d3ydx3 = np.gradient(d2ydx2, x)

print("First derivative at x=5:",dydx[np.where(x == 5)[0][0]])
print("Second derivative at x=5:",d2ydx2[np.where(x == 5)[0][0]])
print("Third derivative at x=5:",d3ydx3[np.where(x == 5)[0][0]])
```

```python
print(x)
print(np.where(x == 5))
print(np.where(x == 5)[0][0])
x[np.where(x == 5)[0][0]]
```

### Série de Taylor

Vamos plotar o polinômio de Taylor $T_3(x)$ de grau 3 centrado em $x=0$ para $f(x) = \frac{3e^x}{x^2 + x + 1}$ sobre o intervalo $x \in [-3,3]$. Primeiro, vamos plotar o gráfico $y=f(x)$:

```python
x = np.linspace(-3,3,100+1)
f = lambda x: 3*np.exp(x) / (x**2 + x + 1)
y = f(x)
plt.plot(x,y);
plt.show()
```

Vamos calcular os coeficientes $a_n = \frac{f^{(n)}(0)}{n!}$ para $n=0,1,2,3$:

```python
# First derivative
dydx = np.gradient(f(x), x)

# Second derivative
d2ydx2 = np.gradient(dydx, x)

# Third derivative
d3ydx3 = np.gradient(d2ydx2, x)


a0 = f(0)
a1 = dydx[np.where(x == 0)[0][0]]
a2 = d2ydx2[np.where(x == 0)[0][0]] / 2
a3 = d3ydx3[np.where(x == 0)[0][0]] / 6
# The parameter order specifies the number of points to use
# The value order must be odd and at least n + 1

print(a0,a1,a2,a3)
```

Finalmente, vamos plotar $f(x)$ e $T_3(x)$ juntos:

```python
T0 = a0+0*x
T1 = a0 + a1*x
T2 = a0 + a1*x + a2*x**2
T3 = a0 + a1*x + a2*x**2 + a3*x**3
plt.plot(x,y,x,T3,x,T2,x,T1,x,T0), plt.xlim([-3,3]), plt.ylim([0,5]);
plt.show()
```

### Comprimento de Arco

Escreva uma função chamada `arc_length` que receba os parâmetros `f`, `a`, `b` e `N` e retorne uma aproximação do comprimento de arco de $f(x)$ de $a$ a $b$

$$
L \approx \int_a^b \sqrt{ 1 + \left( f'(x) \right)^2 } dx
$$

A função usa a regra do trapézio (`scipy.integrate.trapz`) para estimar a integral usando $N+1$ pontos igualmente espaçados de $a$ a $b$ (com valor padrão $N=100$), e a fórmula de diferença central para aproximar $f'(x)$ com passo $h=(b-a)/N$. Note que não podemos usar a fórmula de diferença central nos pontos extremos, pois eles usam valores de $x$ fora do intervalo $[a,b]$ e nossa função pode não estar definida lá.

```python
import scipy.integrate as spi

def arc_length(f,a,b,N=100):
    '''Approximate the arc length of y=f(x) from x=a to x=b.

    Parameters
    ----------
    f : function
        Vectorized function of one variable
    a,b : numbers
        Limits of the interval [a,b]
    N : number
        Subintervals in trapezoid method

    Returns
    -------
    float
        Trapezoid method approximation of the integral
        int_a^b sqrt{1 + (f'(x))^2} dx
        representing the arc length of y=f(x) from x=a to x=b.
    '''
    x = np.linspace(a,b,N+1)
    y = f(x)

    # Compute central difference formula for x_k for 0 < k < N
    h = (b-a)/N
    x_interior = x[1:-1]
    df_interior = (f(x_interior + h) - f(x_interior - h))/(2*h)

    # Use forward/backward difference formula at the endpoints
    df_a = (f(a + h) - f(a))/h
    df_b = (f(b) - f(b - h))/h
    df = np.hstack([[df_a],df_interior,[df_b]])

    # Compute values of the integrand in arc length formula
    y = np.sqrt(1 + df**2)

    # Compute the integral
    L = spi.trapezoid(y,x)

    return L
```

Vamos testar nossa função com uma entrada para a qual conhecemos a saída exata. Por exemplo, o comprimento de arco de $f(x)=x$ de $a=0$ a $b=1$ é $L=\sqrt{2}$ e calculamos

```python
f = lambda x: x
a = 0; b = 1;
arc_length(f,a,b)
```

e comparamos com o valor exato

```python
np.sqrt(2)
```

O comprimento de arco de $f(x)=\sqrt{1 - x^2}$ de $a=0$ a $b=\frac{1}{\sqrt{2}}$ é $L=\frac{\pi}{4}$ e calculamos

```python
f = lambda x: np.sqrt(1 - x**2)
a = 0; b = 1/np.sqrt(2);
arc_length(f,a,b)
```

e comparamos com o valor exato

```python
np.pi/4
```

O comprimento de arco de $f(x)=\frac{2x^{3/2}}{3}$ de $a=0$ a $b=1$ é $L = \frac{2}{3}\left( 2^{3/2} - 1 \right)$ e calculamos

```python
f = lambda x: 2*(x**(3/2))/3
a = 0; b = 1; N = 10**5;
arc_length(f,a,b)
```

e comparamos com o valor exato

```python
(2/3)*(2**(3/2) - 1)
```
