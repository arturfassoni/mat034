### **Problema 1: Implementação e Análise de Erro do Método Runge-Kutta (RK4)**

O Método de Euler é simples, mas acumula erros rapidamente. O Método de Runge-Kutta de 4ª Ordem (RK4) é um dos métodos de passo único mais populares e precisos. Ele obtém uma precisão de 4ª ordem (erro global $O(h^4)$) avaliando a derivada $f(t,y)$ em quatro pontos diferentes dentro de cada passo $h$.

As fórmulas de atualização do RK4 são:
$$y_{n+1} = y_n + \frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4)$$
$$t_{n+1} = t_n + h$$
onde:
$$k_1 = f(t_n, y_n)$$
$$k_2 = f(t_n + \frac{h}{2}, y_n + k_1 \frac{h}{2})$$
$$k_3 = f(t_n + \frac{h}{2}, y_n + k_2 \frac{h}{2})$$
$$k_4 = f(t_n + h, y_n + k_3 h)$$

**Tarefas:**

1.  **Implementação:** Crie uma função `rk4(f, t_span, y0, h)` que receba uma EDO $f(t,y)$, um intervalo de tempo `t_span = [t0, tf]`, uma condição inicial `y0` e um tamanho de passo `h`. A função deve retornar dois arrays NumPy: o vetor de tempo `t` e o vetor de solução `y`.
2.  **Análise de Erro:** Para verificar a ordem de precisão do seu método, faremos uma análise de convergência.
    * Resolva a EDO $y' = y$ com $y(0)=1$ no intervalo $t \in [0, 5]$. A solução analítica é $y(t) = e^t$.
    * Calcule o erro global no tempo final $t=5$ para uma gama de tamanhos de passo, por exemplo: `h_values = [0.5, 0.2, 0.1, 0.05, 0.01]`. O erro é `E(h) = |y_numerico(5) - y_exato(5)|`.
    * Crie um gráfico **log-log** do erro $E(h)$ em função do passo $h$.
    * **Análise:** O método RK4 tem erro $E(h) \approx C \cdot h^4$. Portanto, um gráfico log-log de $E$ vs $h$ deve ter uma inclinação de 4. Verifique visualmente se seu gráfico tem essa inclinação.

### **Problema 2: Modelo Epidemiológico SIR**

O modelo SIR é um dos modelos compartimentais mais fundamentais em epidemiologia. Ele divide a população ($N$) em três grupos:
* **S (Suscetíveis):** Indivíduos que podem contrair a doença.
* **I (Infectados):** Indivíduos que têm a doença e podem transmiti-la.
* **R (Removidos):** Indivíduos que se recuperaram (e ganharam imunidade) ou faleceram.

O modelo é descrito por um sistema de 3 EDOs não lineares:
$$\frac{dS}{dt} = -\frac{\beta S I}{N}$$
$$\frac{dI}{dt} = \frac{\beta S I}{N} - \gamma I$$
$$\frac{dR}{dt} = \gamma I$$

Onde:
* $N$: População total (constante, $N = S + I + R$).
* $\beta$ (beta): Taxa de transmissão (quão rápido a doença se espalha).
* $\gamma$ (gama): Taxa de recuperação (quão rápido os infectados são "removidos").

**Tarefas:**

1.  **Implementação:** Crie uma função `modelo_sir(u, t, N, beta, gamma)` que retorne as derivadas $[dS/dt, dI/dt, dR/dt]$. Note a ordem dos argumentos `(u, t, ...)` necessária para a função `scipy.integrate.odeint`.
2.  **Simulação (Epidemia):**
    * Defina os parâmetros: $N = 1000$, $\beta = 0.35$ (transmissão alta), $\gamma = 0.1$ (recuperação em 10 dias).
    * Condições iniciais: $I_0 = 1$, $R_0 = 0$, $S_0 = N - I_0 - R_0$.
    * Intervalo de tempo: $t = 150$ dias.
    * Use `scipy.integrate.odeint` para resolver o sistema.
    * Plote as curvas de $S(t)$, $I(t)$ e $R(t)$ no mesmo gráfico.
3.  **Análise do Pico:** Para a simulação acima, encontre e imprima o número máximo de infectados (o "pico da epidemia") e o dia em que ele ocorreu.
4.  **Análise do $R_0$:** O Número Básico de Reprodução, $R_0 = \beta / \gamma$, determina se uma epidemia ocorre. Se $R_0 > 1$, a epidemia cresce; se $R_0 < 1$, ela desaparece.
    * Calcule o $R_0$ para a simulação anterior.
    * **Simulação 2 (Contenção):** Faça uma nova simulação com $\beta = 0.05$ (transmissão baixa, ex: distanciamento social) e $\gamma=0.1$. Calcule o novo $R_0$.
    * **Gráfico Comparativo:** Plote as curvas de $I(t)$ (apenas os infectados) das duas simulações no mesmo gráfico. O que você observa?

### **Problema 3: Equações de Lorenz**

As equações de Lorenz são o sistema de equações diferenciais não lineares

\begin{align}
\frac{dx}{dt} &= \sigma(y - x) \\\
\frac{dy}{dt} &= x(\rho - z) - y \\\
\frac{dz}{dt} &= xy - \beta z
\end{align}

onde $\sigma$, $\rho$ e $\beta$ são números positivos.

As Equações de Lorenz são um exemplo clássico de um sistema dinâmico caótico. Além de apenas resolver o sistema, vamos explorar visualmente as características que o tornam tão famoso.

1.  **Função base:** Escreva uma função chamada `lorenz` que receba os parâmetros de entrada `sigma`, `rho`, `beta`, `u0`, `t0`, `tf`, `N` e `plot_vars` (com valor padrão `[0,1]`). A função calcula e plota uma aproximação numérica da solução correspondente das equações de Lorenz usando a função `scipy.integrate.odeint`. Os parâmetros de entrada são:
    * `sigma`, `rho` e `beta` definem os parâmetros $\sigma$, $\rho$ e $\beta$
    * `u0` é uma lista de números de comprimento 3 definindo as condições iniciais $[x(t_0),y(t_0),z(t_0)]$
    * `t0` é o início do intervalo de integração $[t_0,t_f]$
    * `tf` é o fim do intervalo de integração $[t_0,t_f]$
    * `N` é um inteiro especificando o número de pontos igualmente espaçados de $t_0$ a $t_f$ (inclusive) sobre os quais calcular a solução do sistema
    * `plot_vars` é uma lista de comprimento 2 especificando quais 2 componentes plotar, onde $x=0$, $y=1$ e $z=2$. Por exemplo, se `plot_vars` for $[0,1]$, plote a solução $x$ versus $y$. Se `plot_vars` for $[1,2]$, plote a solução $y$ versus $z$. Note que $x$ versus $y$ significa que $x$ é o eixo horizontal e $y$ é o vertical. O valor padrão é `[0,1]`, que plota $x$ versus $y$.
    
    A função `lorenz` retorna um array NumPy 2D com 4 colunas:
    * a coluna no índice 0 é o array de $N$ valores de $t$ igualmente espaçados de $t_0$ a $t_f$ (inclusive)
    * a coluna no índice 1 é o array de valores de $x$ da solução
    * a coluna no índice 2 é o array de valores de $y$ da solução
    * a coluna no índice 3 é o array de valores de $z$ da solução

2.  **Visualização do Atrator 3D:** O "Atrator de Lorenz" é um objeto tridimensional. Use a função `lorenz` que você criou e o kit de ferramentas `mplot3d` do Matplotlib para gerar um gráfico 3D da trajetória.
    * Use os parâmetros clássicos do caos: $\sigma=10$, $\rho=28$, $\beta=8/3$.
    * Condição inicial: `u0 = [0, 1, 1.05]`.
    * Integre de `t0 = 0` a `tf = 40` com `N = 10000` pontos.
    * Plote `z(t)` versus `y(t)` versus `x(t)` para visualizar a famosa forma de "borboleta" do atrator.

3.  **Sensibilidade às Condições Iniciais (O Efeito Borboleta):** Uma marca registrada do caos é a extrema sensibilidade a pequenas perturbações nas condições iniciais. Vamos visualizar isso:
    * **Simulação 1:** Resolva o sistema com os mesmos parâmetros do item anterior e a condição inicial `u0_A = [0, 1, 1.05]`.
    * **Simulação 2:** Crie uma nova condição inicial `u0_B` com uma perturbação minúscula: `u0_B = [0, 1, 1.05001]`. Resolva o sistema novamente com essa nova condição.
    * **Gráfico Comparativo:** Crie um gráfico que mostre **apenas a coordenada x(t) em função do tempo t** para *ambas* as simulações.

3.  **Retratos de Fase (Projeções 2D):** Crie uma única figura com três subplots para visualizar as "sombras" do atrator 3D nos três planos principais:
    * Subplot 1: `y` versus `x`.
    * Subplot 2: `z` versus `x`.
    * Subplot 3: `z` versus `y`.


### **Problema 4: Oscilador Amortecido**

A equação do oscilador amortecido não linear descreve um sistema realista onde o amortecimento depende da velocidade. Vamos analisar seu comportamento no "espaço de fase".

1.  **Função base:** Escreva uma função chamada `damping` que receba os parâmetros de entrada `m`, `b`, `k`, `F`, `u0`, `t0`, `tf` e `N`. A função usa `scipy.integrate.odeint` para calcular uma aproximação numérica da solução correspondente da equação de amortecimento não linear:
    $$ m y'' + b |y'| y' + ky = F(t) $$
    Os parâmetros de entrada são:
    * `m`, `b` e `k` são números positivos na equação de amortecimento não linear
    * `F` é uma função de uma variável $F(t)$ na equação de amortecimento não linear
    * `u0` é uma lista de números de comprimento 2 definindo as condições iniciais $[y(t_0),y'(t_0)]$
    * `t0` é o início do intervalo de integração $[t_0,t_f]$
    * `tf` é o fim do intervalo de integração $[t_0,t_f]$
    * `N` é um inteiro especificando o número de pontos igualmente espaçados de $t_0$ a $t_f$ (inclusive) sobre os quais calcular a solução

    A função `damping` plota a aproximação da solução $y(t)$ e retorna um array Numpy 2D com 2 colunas:
    * a coluna no índice 0 é o array de $N$ valores de $t$ igualmente espaçados de $t_0$ a $t_f$ (inclusive)
    * a coluna no índice 1 é o array de valores de $y$ da solução

2.  **Construção do Retrato de Fase:** O retrato de fase de um oscilador é um gráfico da sua **velocidade ($y'$)** em função da sua **posição ($y$)**. Ele mostra a evolução do estado do sistema ao longo do tempo.
    * Modifique sua função `damping` (ou crie uma nova) para que ela retorne não apenas a posição `y(t)`, mas também a velocidade `y'(t)`. A função `scipy.integrate.odeint` já calcula isso (é a segunda coluna do resultado `U`).
    * Use os parâmetros: `m = 1`, `b = 0.5`, `k = 10`, `F(t) = 0` (sem força externa), e condição inicial `u0 = [3, 0]` (deslocamento inicial de 3, velocidade inicial zero). Integre de `t = 0` a `t = 20`.
    * Crie um gráfico de `y'(t)` (eixo vertical) versus `y(t)` (eixo horizontal).
    * **Análise:** O que a forma da trajetória (uma espiral convergindo para a origem) lhe diz sobre o comportamento de longo prazo do sistema?

3.  **Comparação: Amortecimento Linear vs. Não Linear:** A equação dada tem um amortecimento não linear ($b|y'|y'$). Como isso se compara a um amortecimento linear mais simples ($cy'$)?
    * **Simulação 1 (Não Linear):** Plote a solução `y(t)` em função de `t` para o sistema não linear da tarefa anterior (`m=1, b=0.5, k=10, u0=[3,0]`).
    * **Simulação 2 (Linear):** Resolva a equação do oscilador amortecido **linear** ($my'' + cy' + ky = 0$) com os mesmos parâmetros `m`, `k`, e `u0`. Escolha um coeficiente de amortecimento linear `c` comparável (por exemplo, `c = 1.0`). Plote a solução no mesmo gráfico.
    * **Análise Comparativa:** Compare as duas curvas `y(t)`. Você consegue notar alguma diferença na forma como a amplitude decai? O amortecimento não linear, que é mais forte em altas velocidades e mais fraco em baixas velocidades, produz um decaimento diferente do amortecimento linear.
