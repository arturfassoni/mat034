# Autovalores e Autovetores

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.linalg as la
```

## Definição

Seja $A$ uma matriz quadrada. Um vetor não nulo $\mathbf{v}$ é um [autovetor](https://en.wikipedia.org/wiki/Eigenvalues_and_eigenvectors) para $A$ com [autovalor](https://en.wikipedia.org/wiki/Eigenvalues_and_eigenvectors) $\lambda$ se

$$
A\mathbf{v} = \lambda \mathbf{v}
$$

Rearranjando a equação, vemos que $\mathbf{v}$ é uma solução do sistema homogêneo de equações

$$
\left( A - \lambda I \right) \mathbf{v} = \mathbf{0}
$$

onde $I$ é a matriz identidade de tamanho $n$. Soluções não triviais existem apenas se a matriz $A - \lambda I$ for singular, o que significa que $\mathrm{det}(A - \lambda I) = 0$. Portanto, os autovalores de $A$ são as raízes do [polinômio característico](https://en.wikipedia.org/wiki/Characteristic_polynomial)

$$
p(\lambda) = \mathrm{det}(A - \lambda I)
$$

## scipy.linalg.eig

A função [`scipy.linalg.eig`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.eig.html) calcula os autovalores e autovetores de uma matriz quadrada $A$.

Vamos considerar um exemplo simples com uma matriz diagonal:

```python
A = np.array([[1,0],[0,-2]])
print(A)
```

A função `la.eig` retorna uma tupla `(eigvals,eigvecs)` onde `eigvals` é um array NumPy 1D de números complexos que fornece os autovalores de $A$, e `eigvecs` é um array NumPy 2D com os autovetores correspondentes nas colunas:

```python
results = la.eig(A)
```

Os autovalores de $A$ são:

```python
print(results[0])
```

Os autovetores correspondentes são:

```python
print(results[1])
```

Podemos [desempacotar a tupla](../../python/sequences/#unpacking-a-sequence):

```python
eigvals, eigvecs = la.eig(A)
print(eigvals)
```

```python
print(eigvecs)
```

Se soubermos que os autovalores são números reais (ou seja, se $A$ for simétrica), podemos usar o método `.real` do array NumPy para converter o array de autovalores para números reais:

```python
eigvals = eigvals.real
print(eigvals)
```

Note que a posição de um autovalor no array `eigvals` corresponde à coluna em `eigvecs` com seu autovetor:

```python
lambda1 = eigvals[1]
print(lambda1)
```

```python
v1 = eigvecs[:,1].reshape(2,1)
print(v1)
```

```python
A @ v1
```

```python
lambda1 * v1
```

## Exemplos

### Matrizes Simétricas

Os autovalores de uma [matriz simétrica](https://en.wikipedia.org/wiki/Symmetric_matrix) são sempre reais e os autovetores são sempre ortogonais! Vamos verificar esses fatos com algumas matrizes aleatórias:

```python
n = 4
P = np.random.randint(0,10,(n,n))
print(P)
```

Crie a matriz simétrica $S = P P^T$:

```python
S = P @ P.T
print(S)
```

Vamos desempacotar os autovalores e autovetores de $S$:

```python
evals, evecs = la.eig(S)
print(evals)
```

Os autovalores têm todos parte imaginária nula e, portanto, são de fato números reais:

```python
evals = evals.real
print(evals)
```

Os autovetores correspondentes de $S$ são:

```python
print(evecs)
```

Vamos verificar se os autovetores são ortogonais entre si:

```python
v1 = evecs[:,0] # A primeira coluna é o primeiro autovetor
print(v1)
```

```python
v2 = evecs[:,1] # A segunda coluna é o segundo autovetor
print(v2)
```

```python
v1 @ v2
```

O produto escalar dos autovetores $\mathbf{v}_1$ e $\mathbf{v}_2$ é zero (o número acima é *muito* próximo de zero e se deve a erros de arredondamento nos cálculos) e, portanto, eles são ortogonais!

Podemos verificar a ortogonalidade de todos os pares de uma só vez: se as colunas de `evecs` são ortonormais, então $V^TV = I$.

```python
np.round(evecs.T @ evecs, 10)
```

Para matrizes simétricas, é melhor usar a função [`scipy.linalg.eigh`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.eigh.html): ela usa um algoritmo próprio para matrizes simétricas (mais rápido e preciso), retorna autovalores **reais** em ordem crescente e autovetores ortonormais.

```python
evals, evecs = la.eigh(S)
print(evals)
```

### Autovalores Complexos

Matrizes reais que não são simétricas podem ter autovalores complexos. Por exemplo, a matriz de rotação por um ângulo $\theta$,

$$
R_\theta = \begin{bmatrix} \cos\theta & -\sin\theta \\\ \sin\theta & \cos\theta \end{bmatrix},
$$

não deixa nenhuma direção do plano invariante (para $\theta \neq 0, \pi$), logo não tem autovetores reais. Seus autovalores são $\lambda = \cos\theta \pm i\sin\theta = e^{\pm i\theta}$, com $|\lambda| = 1$.

```python
theta = np.pi/6
R = np.array([[np.cos(theta),-np.sin(theta)],[np.sin(theta),np.cos(theta)]])
evals, evecs = la.eig(R)
print("Autovalores:", evals)
print("Módulos:", np.abs(evals))
print("Argumentos (graus):", np.degrees(np.angle(evals)))
```

### Diagonalização

Uma matriz quadrada $M$ é [diagonalizável](https://en.wikipedia.org/wiki/Diagonalizable_matrix) se for semelhante a uma matriz diagonal. Em outras palavras, $M$ é diagonalizável se existir uma matriz invertível $P$ tal que $D = P^{-1}MP$ seja uma matriz diagonal.

Um belo resultado em álgebra linear é que uma matriz quadrada $M$ de tamanho $n$ é diagonalizável se, e somente se, $M$ tiver $n$ autovetores independentes. Além disso, $M = PDP^{-1}$ onde as colunas de $P$ são os autovetores de $M$ e $D$ tem os autovalores correspondentes ao longo da diagonal.

Vamos usar isso para construir uma matriz com autovalores dados $\lambda_1 = 3, \lambda_2 = 1$, e autovetores $v_1 = [1,1]^T, v_2 = [1,-1]^T$.

```python
P = np.array([[1,1],[1,-1]])
print(P)
```

```python
D = np.diag((3,1))
print(D)
```

```python
M = P @ D @ la.inv(P)
print(M)
```

Vamos verificar se os autovalores de $M$ são 3 e 1:

```python
evals, evecs = la.eig(M)
print(evals)
```

Verifique os autovetores:

```python
print(evecs)
```

### Potências de Matrizes

Seja $M$ uma matriz quadrada. Calcular potências de $M$ por multiplicação de matrizes

$$
M^k = \underbrace{M M \cdots M}_k
$$

é computacionalmente caro. Em vez disso, vamos usar a diagonalização para calcular $M^k$ de forma mais eficiente

$$
M^k = \left( P D P^{-1} \right)^k = \underbrace{P D P^{-1} P D P^{-1} \cdots P D P^{-1}}_k = P D^k P^{-1}
$$

Vamos calcular $M^{20}$ de ambas as formas e comparar o tempo de execução.

```python
Pinv = la.inv(P)
```

```python
k = 20
```

```python
%%timeit
result = M.copy()
for _ in range(1,k):
    result = result @ M
```

Vamos usar a diagonalização para fazer o mesmo cálculo.

```python
%%timeit
P @ D**k @ Pinv     # D**k eleva cada elemento a k: funciona porque D é diagonal
```

A diagonalização calcula $M^{k}$ muito mais rápido! Atenção: o operador `**` é elemento a elemento; `D**k` só coincide com a potência matricial $D^k$ porque $D$ é diagonal.

## Método das Potências

Para matrizes grandes, calcular todos os autovalores pode ser caro, e muitas vezes só nos interessa o **autovalor dominante** (o de maior módulo) e seu autovetor. É o caso do crescimento de populações, das cadeias de Markov e do PageRank do Google.

Suponha que $A$ seja diagonalizável, com autovalores $|\lambda_1| > |\lambda_2| \geq \cdots \geq |\lambda_n|$ e autovetores $\mathbf{v}_1, \dots, \mathbf{v}_n$. Escrevendo um vetor inicial como $\mathbf{x}_0 = c_1\mathbf{v}_1 + \cdots + c_n\mathbf{v}_n$ (com $c_1 \neq 0$), temos

$$
A^k\mathbf{x}_0 = c_1\lambda_1^k\mathbf{v}_1 + \cdots + c_n\lambda_n^k\mathbf{v}_n = \lambda_1^k\left(c_1\mathbf{v}_1 + c_2\left(\frac{\lambda_2}{\lambda_1}\right)^k\mathbf{v}_2 + \cdots\right).
$$

Como $|\lambda_j/\lambda_1| < 1$, os termos entre parênteses (exceto o primeiro) tendem a zero: **$A^k\mathbf{x}_0$ aponta cada vez mais na direção de $\mathbf{v}_1$**. A velocidade de convergência é determinada pela razão $|\lambda_2/\lambda_1|$.

O **método das potências** consiste em multiplicar repetidamente por $A$, normalizando o vetor a cada passo (para evitar que ele cresça ou encolha demais):

$$
\mathbf{x}_{k+1} = \frac{A\mathbf{x}_k}{\|A\mathbf{x}_k\|}.
$$

O autovalor é estimado pelo [quociente de Rayleigh](https://en.wikipedia.org/wiki/Rayleigh_quotient) $\lambda \approx \mathbf{x}_k \cdot A\mathbf{x}_k$ (com $\|\mathbf{x}_k\| = 1$).

```python
def potencias(A,x0,tol=1e-10,maxit=1000):
    """Método das potências: retorna (autovalor dominante, autovetor, número de iterações)."""
    x = np.array(x0,dtype=float)
    x = x/la.norm(x)
    lam = x @ A @ x
    for k in range(1,maxit+1):
        y = A @ x
        x = y/la.norm(y)
        lam_novo = x @ A @ x
        if abs(lam_novo - lam) < tol:
            return lam_novo, x, k
        lam = lam_novo
    print("Atenção: o método não convergiu.")
    return lam, x, maxit
```

```python
A = np.array([[2,1,0],[1,3,1],[0,1,4]])
lam, v, k = potencias(A,[1,1,1])
print("Método das potências:", lam, "em", k, "iterações")
print("la.eigh:             ", la.eigh(A)[0])
```

Vamos observar a convergência: o erro na direção do autovetor deve diminuir aproximadamente como $|\lambda_2/\lambda_1|^k$, o que aparece como uma reta em escala semilogarítmica.

```python
evals, evecs = la.eigh(A)
v1 = evecs[:,-1]                    # autovetor do maior autovalor
razao = abs(evals[-2]/evals[-1])
x = np.array([1.,1.,1.]); x = x/la.norm(x)
erros = []
for k in range(40):
    x = A @ x; x = x/la.norm(x)
    erros.append(min(la.norm(x - v1), la.norm(x + v1)))   # o autovetor é definido a menos de sinal

plt.semilogy(erros,'b.-',label="erro do método das potências")
plt.semilogy(erros[0]*razao**np.arange(40),'r--',label=r"$|\lambda_2/\lambda_1|^k$")
plt.xlabel("iteração $k$"), plt.ylabel("erro"), plt.grid(True), plt.legend()
plt.title("Convergência do método das potências")
plt.show()
```
