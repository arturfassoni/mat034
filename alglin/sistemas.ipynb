## Sistemas Lineares

Um [sistema de equações lineares](https://en.wikipedia.org/wiki/System_of_linear_equations) é um conjunto de equações lineares

\begin{align}
a_{0,0}x_0 + a_{0,1}x_1 + \cdots + a_{0,n}x_n & = b_0 \\\
a_{1,0}x_0 + a_{1,1}x_1 + \cdots + a_{1,n}x_n & = b_1 \\\
& \vdots \\\
a_{m,0}x_0 + a_{m,1}x_1 + \cdots + a_{m,n}x_n & = b_m \\\
\end{align}

Em notação matricial, um sistema linear é $A \mathbf{x}= \mathbf{b}$, onde

$$
A = \begin{bmatrix}
a_{0,0} & a_{0,1} & \cdots & a_{0,n} \\\
a_{1,0} & a_{1,1} & \cdots & a_{1,n} \\\
\vdots & & & \vdots \\\
a_{m,0} & a_{m,1} & \cdots & a_{m,n} \\\
\end{bmatrix}
 \ \ , \ \
\mathbf{x} = \begin{bmatrix}
x_0 \\\ x_1 \\\ \vdots \\\ x_n
\end{bmatrix}
 \ \ , \ \
\mathbf{b} = \begin{bmatrix}
b_0 \\\ b_1 \\\ \vdots \\\ b_m
\end{bmatrix}
$$

## Eliminação Gaussiana

O procedimento geral para resolver um sistema de equações lineares é chamado de [eliminação gaussiana](https://en.wikipedia.org/wiki/Gaussian_elimination). A ideia é realizar operações elementares nas linhas para reduzir o sistema à sua forma escalonada e então resolvê-lo.

### Operações Elementares nas Linhas

As [operações elementares nas linhas](https://en.wikipedia.org/wiki/Elementary_matrix#Elementary_row_operations) incluem:

1. Somar $k$ vezes a linha $j$ à linha $i$.
2. Multiplicar a linha $i$ pelo escalar $k$.
3. Trocar as linhas $i$ e $j$.

Cada uma das operações elementares nas linhas é o resultado da multiplicação por uma matriz elementar (à esquerda).
Para somar $k$ vezes a linha $j$ à linha $i$ em uma matriz $A$, multiplicamos $A$ pela matriz $E$, onde $E$ é igual à matriz identidade, exceto que o elemento $i,j$ é $E_{i,j} = k$. Por exemplo, se $A$ é 3x3 e queremos somar 3 vezes a linha 2 à linha 0 (usando indexação baseada em 0), então

$$
E_1 = \begin{bmatrix}
1 & 0 & 3 \\\
0 & 1 & 0 \\\
0 & 0 & 1
\end{bmatrix}
$$

Vamos verificar o cálculo:

```python
A = np.array([[1,1,2],[-1,3,1],[0,5,2]])
print(A)
```

```python
E1 = np.array([[1,0,3],[0,1,0],[0,0,1]])
print(E1)
```

```python
E1 @ A
```

Para multiplicar a linha $i$ por $k$ em uma matriz $A$, multiplicamos $A$ pela matriz $E$, onde $E$ é igual à matriz identidade, exceto que o elemento $i,i$ é $E_{i,i} = k$. Por exemplo, se $A$ é 3x3 e queremos multiplicar a linha 1 por -2, então

$$
E_2 = \begin{bmatrix}
1 & 0 & 0 \\\
0 & -2 & 0 \\\
0 & 0 & 1
\end{bmatrix}
$$

Vamos verificar o cálculo:

```python
E2 = np.array([[1,0,0],[0,-2,0],[0,0,1]])
print(E2)
```

```python
E2 @ A
```

Finalmente, para trocar a linha $i$ e a linha $j$ em uma matriz $A$, multiplicamos $A$ pela matriz $E$, onde $E$ é igual à matriz identidade, exceto que $E_{i,i} = 0$, $E_{j,j} = 0$, $E_{i,j} = 1$ e $E_{j,i} = 1$. Por exemplo, se $A$ é 3x3 e queremos trocar a linha 1 e a linha 2, então

$$
E_3 = \begin{bmatrix}
1 & 0 & 0 \\\
0 & 0 & 1 \\\
0 & 1 & 0
\end{bmatrix}
$$

Vamos verificar o cálculo:

```python
E3 = np.array([[1,0,0],[0,0,1],[0,1,0]])
print(E3)
```

```python
E3 @ A
```

### Implementação

Vamos escrever funções para implementar as operações elementares nas linhas. Primeiramente, vamos escrever uma função chamada `add_row` que recebe os parâmetros de entrada $A$, $k$, $i$ e $j$ e retorna o array NumPy resultante de somar $k$ vezes a linha $j$ à linha $i$ na matriz $A$. Se $i=j$, vamos dizer que a função escala a linha $i$ por $k+1$, pois este seria o resultado de $k$ vezes a linha $i$ somada à linha $i$.

```python
def add_row(A,k,i,j):
    """Soma k vezes a linha j à linha i na matriz A."""
    n = A.shape[0]
    E = np.eye(n)
    if i == j:
        E[i,i] = k + 1
    else:
        E[i,j] = k
    return E @ A
```

Vamos testar nossa função:

```python
M = np.array([[1,1],[3,2]])
print(M)
```

```python
add_row(M,2,0,1)
```

```python
add_row(M,3,1,1)
```

Vamos escrever uma função chamada `scale_row` que recebe 3 parâmetros de entrada $A$, $k$ e $i$ e retorna a matriz que resulta da multiplicação da linha $i$ por $k$ na matriz $A$.

```python
def scale_row(A,k,i):
    """Multiplica a linha i por k na matriz A."""
    n = A.shape[0]
    E = np.eye(n)
    E[i,i] = k
    return E @ A
```

```python
M = np.array([[3,1],[-2,7]])
print(M)
```

```python
scale_row(M,3,1)
```

```python
A = np.array([[1,1,1],[1,-1,0]])
print(A)
```

```python
scale_row(A,5,1)
```

Vamos escrever uma função chamada `switch_rows` que recebe 3 parâmetros de entrada $A$, $i$ e $j$ e retorna a matriz que resulta da troca das linhas $i$ e $j$ na matriz $A$.

```python
def switch_rows(A,i,j):
    """Troca as linhas i e j na matriz A."""
    n = A.shape[0]
    E = np.eye(n)
    E[i,i] = 0
    E[j,j] = 0
    E[i,j] = 1
    E[j,i] = 1
    return E @ A
```

```python
A = np.array([[1,1,1],[1,-1,0]])
print(A)
```

```python
switch_rows(A,0,1)
```

### Implementação do Algoritmo

Fazer as operações à mão, uma a uma, ajuda a entender o método, mas o objetivo é automatizá-lo. A eliminação gaussiana tem duas etapas:

1. **Eliminação:** para cada coluna $k = 0, \dots, n-2$, zeramos os elementos abaixo do pivô $a_{kk}$, subtraindo da linha $i$ o múltiplo $m_{ik} = a_{ik}/a_{kk}$ da linha $k$.
2. **Substituição regressiva:** com o sistema triangular superior $U\mathbf{x} = \mathbf{c}$, calculamos

$$
x_{n-1} = \frac{c_{n-1}}{u_{n-1,n-1}}, \qquad x_i = \frac{1}{u_{ii}}\left(c_i - \sum_{j=i+1}^{n-1} u_{ij}x_j\right), \quad i = n-2, \dots, 0.
$$

Para evitar divisões por pivôs nulos ou muito pequenos (que amplificam os erros de arredondamento), usamos o **pivoteamento parcial**: antes de eliminar a coluna $k$, trocamos a linha $k$ pela linha (abaixo dela) que tem o maior $|a_{ik}|$.

```python
def gauss(A,b):
    """Resolve Ax = b por eliminação gaussiana com pivoteamento parcial."""
    A = np.array(A,dtype=float)            # cópias, para não alterar os originais
    b = np.array(b,dtype=float).flatten()
    n = len(b)
    # 1. Eliminação
    for k in range(n-1):
        p = k + np.argmax(np.abs(A[k:,k]))  # linha do maior pivô
        if p != k:
            A[[k,p]] = A[[p,k]]
            b[[k,p]] = b[[p,k]]
        for i in range(k+1,n):
            m = A[i,k]/A[k,k]
            A[i,k:] = A[i,k:] - m*A[k,k:]
            b[i] = b[i] - m*b[k]
    # 2. Substituição regressiva
    x = np.zeros(n)
    for i in range(n-1,-1,-1):
        x[i] = (b[i] - A[i,i+1:] @ x[i+1:])/A[i,i]
    return x
```

```python
A = np.array([[6,15,1],[8,7,12],[2,7,8]])
b = np.array([2,14,10])
print(gauss(A,b))
print(la.solve(A,b))
```

Contando as operações dos laços, a eliminação gasta cerca de $\frac{2}{3}n^3$ operações aritméticas, enquanto a substituição regressiva gasta cerca de $n^2$. Dobrar o tamanho do sistema multiplica o custo por aproximadamente 8.

## `scipy.linalg.solve`

Estamos interessados principalmente em sistemas lineares $A \mathbf{x} = \mathbf{b}$ onde há uma solução única $\mathbf{x}$. Este é o caso quando $A$ é uma matriz quadrada ($m=n$) e $\mathrm{det}(A) \not= 0$. Para resolver tal sistema, podemos usar a função [`scipy.linalg.solve`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.solve.html).

A função retorna uma solução do sistema de equações $A \mathbf{x} = \mathbf{b}$. Por exemplo:

```python
A = np.array([[1,1],[1,-1]])
print(A)
```

```python
b1 = np.array([2,0])
print(b1)
```

E resolvemos:

```python
x1 = la.solve(A,b1)
print(x1)
```

Note que a saída $\mathbf{x}$ é retornada como um array NumPy 1D quando o vetor $\mathbf{b}$ (o lado direito) é inserido como um array NumPy 1D. Se inserirmos $\mathbf{b}$ como um array NumPy 2D, então a saída é um array NumPy 2D. Por exemplo:

```python
A = np.array([[1,1],[1,-1]])
b2 = np.array([2,0]).reshape(2,1)
x2 = la.solve(A,b2)
print(x2)
```

Finalmente, se o lado direito $\mathbf{b}$ é uma matriz, então a saída é uma matriz do mesmo tamanho. É a solução de $A \mathbf{X} = \mathbf{B}$ quando $\mathbf{B}$ é uma matriz. Por exemplo:

```python
A = np.array([[1,1],[1,-1]])
B = np.array([[2,2],[0,1]])
X = la.solve(A,B)
print(X)
```

```python
Ainv = la.solve(A,np.eye(2))
print(Ainv)
print(Ainv @ A)
```

## Decomposição LU

A eliminação gaussiana pode ser registrada como uma fatoração da matriz. Os multiplicadores $m_{ik}$ formam uma matriz triangular inferior $L$ (com 1 na diagonal), o resultado da eliminação é uma matriz triangular superior $U$, e as trocas de linha formam uma matriz de permutação $P$:

$$
A = PLU.
$$

A função `scipy.linalg.lu` calcula essa fatoração:

```python
A = np.array([[6,15,1],[8,7,12],[2,7,8]])
P, L, U = la.lu(A)
print(L); print(U)
print(P @ L @ U)
```

A vantagem da fatoração é que, uma vez calculadas $L$ e $U$ (custo $\approx \frac{2}{3}n^3$), resolver $A\mathbf{x} = \mathbf{b}$ exige apenas duas substituições triangulares (custo $\approx 2n^2$). Isso é muito útil quando precisamos resolver **vários sistemas com a mesma matriz** e lados direitos diferentes. No SciPy, usamos `la.lu_factor` (fatora uma vez) e `la.lu_solve` (resolve para cada $\mathbf{b}$):

```python
N = 500
A = np.random.rand(N,N)
lu, piv = la.lu_factor(A)          # fatoração: feita uma única vez
for k in range(3):
    b = np.random.rand(N)
    x = la.lu_solve((lu,piv),b)    # cada solução é barata
    print("Resíduo ||Ax - b|| =", la.norm(A @ x - b))
```

## Condicionamento

Em um computador, os dados de um sistema linear sempre têm pequenos erros (de medição ou de arredondamento). Uma pergunta fundamental é: **quanto um pequeno erro em $\mathbf{b}$ altera a solução $\mathbf{x}$?** A resposta é dada pelo **número de condição**

$$
\kappa(A) = \|A\|\,\|A^{-1}\|,
$$

que satisfaz

$$
\frac{\|\Delta \mathbf{x}\|}{\|\mathbf{x}\|} \leq \kappa(A)\,\frac{\|\Delta \mathbf{b}\|}{\|\mathbf{b}\|}.
$$

Sempre temos $\kappa(A) \geq 1$. Se $\kappa(A)$ é pequeno, a matriz é **bem condicionada**; se é grande, é **mal condicionada**, e pequenos erros nos dados podem produzir grandes erros na solução. Uma regra prática: com aritmética de 16 dígitos, perdemos cerca de $\log_{10}\kappa(A)$ dígitos de precisão. Calculamos $\kappa(A)$ com `np.linalg.cond`.

Um exemplo clássico de matriz mal condicionada é a [matriz de Hilbert](https://en.wikipedia.org/wiki/Hilbert_matrix), $H_{ij} = \frac{1}{i+j+1}$. Vamos resolver $H\mathbf{x} = \mathbf{b}$ com $\mathbf{b} = H\mathbf{1}$, cuja solução exata é $\mathbf{x} = \mathbf{1}$ (o vetor de uns):

```python
for n in [4,8,12,16]:
    H = la.hilbert(n)
    x_exato = np.ones(n)
    b = H @ x_exato
    x = la.solve(H,b)
    print(f"n = {n:2d}   cond(H) = {np.linalg.cond(H):.1e}   erro relativo = {la.norm(x - x_exato)/la.norm(x_exato):.1e}")
```

Para $n = 12$, o número de condição já é da ordem de $10^{16}$ e a solução calculada tem erro de cerca de 20%; para $n = 16$, o erro é maior que a própria solução. O SciPy chega a emitir um aviso (`LinAlgWarning: An ill-conditioned matrix detected`), mas `la.solve` está funcionando perfeitamente: O problema não está no algoritmo, mas na própria matriz. Veremos esse fenômeno novamente na interpolação polinomial com matrizes de Vandermonde.
