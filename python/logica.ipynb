## Valores Booleanos

O tipo [booleano](https://docs.python.org/3/library/stdtypes.html#truth-value-testing) tem apenas dois valores: `True` e `False`. Vamos atribuir um valor booleano a uma variável e verificar o tipo usando a função embutida `type()`:

```python
python_is_fun = True
print(python_is_fun)
```

```python
type(python_is_fun)
```

```python
python_is_fun = 1+2j
type(python_is_fun)
```

Vamos atribuir o valor `False` a uma variável e verificar novamente o tipo:

```python
math_is_scary = False
print(math_is_scary)
```

```python
type(math_is_scary)
```

## Operadores de Comparação

[Operadores de comparação](https://docs.python.org/3/library/stdtypes.html#comparisons) produzem valores Booleanos como saída. Por exemplo, se tivermos variáveis `x` e `y` com valores numéricos, podemos avaliar a expressão `x < y` e o resultado é um valor booleano, `True` ou `False`.

| Operador de Comparação | Descrição  |
| :---: | :---: |
| `<` | estritamente menor que |
| `<=` | menor que ou igual a |
| `>` | estritamente maior que |
| `>=` | maior que ou igual a |
| `==` | igual a |
| `!=` | diferente de |

Por exemplo:

```python
1 == 2
```

```python
1 < 2
```

```python
2 == 2
```

```python
3 != 3.14159
```

```python
20.00000001 >= 20
```

## Operadores Booleanos

Combinamos expressões lógicas usando os [operadores booleanos](https://docs.python.org/3/library/stdtypes.html#boolean-operations-and-or-not) `and`, `or` e `not`.

| Operador Booleano | Descrição |
| :---: | :---: |
| `A and B` | retorna `True` se tanto `A` quanto `B` forem `True` |
| `A or B` | retorna `True` se `A` ou `B` for `True`
| `not A` | retorna `True` se `A` for `False`

Por exemplo:

```python
(1 < 2) and (3 != 5)
```

```python
(1 < 2) and (3 < 1)
```

```python
(1 < 2) or (3 < 1)
```

```python
not (1000 <= 999)
```

## Declarações if

Uma [declaração if](https://docs.python.org/3/tutorial/controlflow.html#if-statements) consiste em um ou mais blocos de código, de modo que apenas um bloco é executado dependendo das expressões lógicas.

Por exemplo, determine se as raízes da equação polinomial $ax^2 + bx + c = 0$ são reais, repetidas ou complexas usando a fórmula de Bhaskara

$$
x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
$$

```python
a = 1
b = -4
c = 1
discriminant = b**2 - 4*a*c
if discriminant > 0:
    print("Discriminant =", discriminant)
    print("Roots are real and distinct.")
elif discriminant < 0: # elif=else if="senão, então se"
    print("Discriminant =", discriminant)
    print("Roots are complex.")
else: #else = "senão"
    print("Discriminant =", discriminant)
    print("Roots are real and repeated.")
```

Os principais pontos a serem observados são:

1. Comece com a palavra-chave `if`.
2. Escreva uma expressão lógica (que retorne `True` ou `False`).
3. Termine a linha com dois pontos `:`.
4. Indente o bloco com 4 espaços após a declaração `if`.
5. Inclua as declarações `elif` e `else`, se necessário.
6. Apenas um dos blocos `if`, `elif` e `else` é executado.
7. O bloco que segue uma declaração `else` será executado apenas se todas as outras expressões lógicas anteriores forem `False`.

### Matriz Invertível

Represente uma matriz quadrada 2 por 2 como uma lista de listas. Por exemplo, represente a matriz

$$
\begin{bmatrix} 2 & -1 \\\ 5 & 7 \end{bmatrix}
$$

como a lista de listas `[[2,-1],[5,7]]`.

Escreva uma função chamada `invertible` que receba um parâmetro de entrada `M`, uma lista de listas representando uma matriz 2 por 2, e retorne `True` se a matriz `M` for invertível e `False` caso contrário.

```python
def invertible(M):
    "Determine if the matrix M = [[a,b],[c,d]] is invertible."
    determinant = M[0][0] * M[1][1] - M[0][1] * M[1][0]
    if determinant != 0:
        return True
    else:
        return False
```

Vamos testar nossa função:

```python
invertible([[1,2],[3,4]])
```

```python
invertible([[1,1],[3,3]])
```

### Concavidade de um Polinômio

Escreva uma função chamada `concave_up` que receba os parâmetros de entrada `p` e `a`, onde `p` é uma lista que representa um polinômio $p(x)$ e `a` é um número, e retorne `True` se a função $p(x)$ tiver concavidade para cima em $x=a$ (ou seja, sua segunda derivada é positiva em $x=a$, $p''(a) > 0$).

Usaremos o teste da segunda derivada para polinômios. Em particular, se tivermos um polinômio de grau $n$

$$
p(x) = c_0 + c_1 x + c_2 x^2 + \cdots + c_n x^n
$$

então a segunda derivada de $p(x)$ em $x=a$ é a soma

$$
p''(a) = 2(1) c_2 + 3(2)c_3 a + 4(3)c_4 a^2 + \cdots + n(n-1)c_n a^{n-2}
$$

```python
def concave_up(p,a):
    "Determine if the polynomial p(x) is concave up at x=a."
    degree = len(p) - 1
    if degree < 2:
        return False
    else:
        # Compute the second derivative p''(a)
        DDp_a = sum([k*(k-1)*p[k]*a**(k-2) for k in range(2,degree + 1)])
        if DDp_a > 0:
            return True
        else:
            return False
```

Vamos testar nossa função em $p(x) = 1 + x - x^3$ em $x=2$. Como $p''(x) = -6x$ e $p''(2) = -12 < 0$, o polinômio tem concavidade para baixo em $x=2$.

```python
p = [1,1,0,-1]
a = 2
concavity = concave_up(p,a)
print(concavity)
```
