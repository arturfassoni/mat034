# Funções



Uma [função](https://docs.python.org/3/tutorial/controlflow.html#defining-functions) recebe parâmetros de entrada, executa uma série de cálculos com essas entradas e, em seguida, retorna um valor de saída final. As funções nos dão uma maneira eficiente de salvar e reutilizar um bloco de código repetidamente com diferentes valores de entrada. Existem funções embutidas na biblioteca padrão do Python e podemos definir nossas próprias funções.

## Funções Embutidas

A biblioteca padrão do Python tem uma coleção de [funções embutidas](https://docs.python.org/3/library/functions.html) prontas para usarmos. Já vimos algumas dessas funções em seções anteriores, como `type()`, `print()` e `sum()`. A seguir, uma lista das funções embutidas que usaremos com mais frequência:

| Função | Descrição |
| ---: | :--- |
| `print(objeto)` | imprime o `objeto` na saída |
| `type(objeto)` | retorna o tipo do `objeto` |
| `abs(x)` | retorna o valor absoluto de `x` (ou módulo se `x` for complexo) |
| `int(x)` | retorna o inteiro construído a partir do float `x` truncando a parte decimal |
| `len(sequencia)` | retorna o comprimento da `sequencia` |
| `sum(sequencia)` | retorna a soma das entradas da `sequencia` |
| `max(sequencia)` | retorna o valor máximo na `sequencia` |
| `min(sequencia)` | retorna o valor mínimo na `sequencia` |

| `list(sequencia)` | retorna uma lista construída a partir da `sequencia` |
| `sorted(sequencia)` | retorna a lista ordenada a partir dos itens em `sequencia` |

| `enumerate(sequencia)` | retorna o objeto enumerate construído a partir da `sequencia` |

## Definindo Funções

Vamos começar com um exemplo simples. Defina uma função que retorna a média de uma sequência de números:

````python
def average(x):
    "Compute the average of the values in the sequence x."
    sum_x = sum(x)
    length_x = len(x)
    return sum_x / length_x
````

````python
x=[1,2,3,4,5]
average(x)
````

Os principais pontos a serem observados são:

1. Comece a definição da função com a palavra-chave `def`.
2. Siga `def` com o nome da função.
3. Siga o nome da função com a lista de parâmetros de entrada separados por vírgulas e dentro de parênteses.
4. Termine a declaração `def` com dois pontos `:`.
5. Indente o corpo da função com 4 espaços.
6. Use a palavra-chave `return` para especificar a saída da função (mas nem sempre é necessário).
7. A segunda linha é uma [string de documentação](https://docs.python.org/3/tutorial/controlflow.html#documentation-strings) (docstring, entre aspas " ... ") que descreve a função.

Em Python, blocos de código são definidos usando [indentação](https://www.python.org/dev/peps/pep-0008/#indentation). Isso significa que linhas de código com a mesma quantidade de indentação são consideradas um bloco. No exemplo acima, as quatro linhas indentadas abaixo da declaração `def` formam o corpo da função.

Observe que não há saída quando executamos a célula que contém a definição da função. Isso ocorre porque apenas definimos a função e ela está esperando que a usemos! Precisamos chamar a função com valores para os parâmetros de entrada, e então a função irá calcular e retornar o valor de saída.

Vamos testar nossa função:

````python
average([1,2,3,4])
````

````python
average(range(1,10))
````

A função retorna o valor esperado. Sucesso!

## Argumentos de Palavra-chave (Keyword Arguments)

Quando definimos funções, listamos os parâmetros de entrada. Estes são chamados de parâmetros posicionais (ou argumentos posicionais) porque a posição na declaração `def` determina qual parâmetro é qual.

````python
def poly(x,y):
    "Compute x + y**2."
    return x + y**2
````

````python
poly(1,2)
````

````python
poly(2,1)
````

Um [argumento de palavra-chave](https://docs.python.org/3/tutorial/controlflow.html#keyword-arguments) nos permite inserir valores padrão para alguns parâmetros e podemos chamá-los pelo nome, e a ordem não importa.

Por exemplo, vamos escrever uma função chamada `psum` que recebe os números `x` e `y` e o argumento de palavra-chave `p` (com valor padrão `p=1`) e retorna a soma das potências

$$
x^p + y^p
$$

````python
def psum(x,y,p=2):
    return x**p + y**p
````

````python
psum(1,2)
````

````python
psum(1,2,p=1)
````

````python
psum(2,3,p=3)
````

Nesta função, `x` e `y` são argumentos posicionais e `p` é um argumento de palavra-chave.

````python
def psum2(x,y,p=2,q=3):
    "uma função com dois parâmetros"
    return x**p + y**q
````

````python
psum2(1,2,q=-1)
````

## Funções Lambda

[Funções Lambda](https://docs.python.org/3/tutorial/controlflow.html#lambda-expressions) são pequenas funções anônimas criadas com a palavra-chave `lambda`:

```python
nome_da_funcao = lambda parametro: valor_de_retorno
```

A expressão do lado direito do operador de atribuição `=` é uma função lambda anônima e nós a atribuímos ao nome de variável `nome_da_funcao`. As funções Lambda são úteis de pelo menos duas maneiras:

* Definir uma função curta em uma única linha de código
* Definir uma função *dentro* de alguma outra expressão Python

Por exemplo, a função de média definida acima pode ser escrita em uma única linha concisa:

````python
average = lambda x: sum(x)/len(x)
````

A palavra-chave `lambda` indica que o que se segue é uma definição de função. O nome da variável `x` antes dos dois pontos é o parâmetro de entrada e a expressão após os dois pontos é o valor de retorno. Vamos verificar se nossa função lambda retorna os valores corretos:

````python
average([1,2,3,4])
````

````python
average([1,2,1,2])
````

````python
custo = lambda x: 1+2*x
custo(3)
````

Crie funções lambda com vários parâmetros de entrada listando nomes de variáveis separados por vírgulas. Por exemplo, vamos criar uma função chamada `hipotenusa` que recebe os parâmetros de entrada `x` e `y` e retorna o comprimento da [hipotenusa](https://en.wikipedia.org/wiki/Hypotenuse) do triângulo retângulo com lados `x` e `y`.

````python
hypotenuse = lambda x,y: (x**2 + y**2)**0.5
````

Verifique se nossa função retorna os valores corretos:

### Função Zeta de Riemann

A [função zeta de Riemann](https://en.wikipedia.org/wiki/Riemann_zeta_function) é a soma infinita

$$
\zeta(s) = \sum_{n = 1}^{\infty} \frac{1}{n^s}
$$

Escreva uma função chamada `zeta` que receba 2 parâmetros de entrada `s` e `N` e retorne a soma parcial:

$$
\sum_{n=1}^N \frac{1}{n^s}
$$

````python
def zeta(s,N):
    "Compute the Nth partial sum of the zeta function at s."
    terms = [1/n**s for n in range(1,N+1)]
    partial_sum = sum(terms)
    return partial_sum
````

Vamos testar nossa função com valores de entrada para os quais conhecemos o resultado:

````python
zeta(1,1)
````

````python
zeta(2,2)
1+(1/2)**2
````

Agora vamos usar nossa função para aproximar [valores especiais da função zeta de Riemann](https://en.wikipedia.org/wiki/Riemann_zeta_function#Specific_values):

$$
\zeta(2) = \frac{\pi^2}{6} \hspace{10mm} \text{e} \hspace{10mm} \zeta(4) = \frac{\pi^4}{90}
$$

Calcule a soma parcial para $s=2$ e $N=100000$:

### Média Harmônica

Escreva uma função chamada `harmonic_mean` que receba um parâmetro de entrada `s`, uma lista de números $x_1, \dots, x_n$ de comprimento $n$, e retorne a [média harmônica](https://en.wikipedia.org/wiki/Harmonic_mean) da sequência:

$$
\frac{n}{\frac{1}{x_1} + \frac{1}{x_2} + \cdots + \frac{1}{x_n}}
$$

````python
def harmonic_mean(s):
    "Compute the harmonic mean of the numbers in the sequence s."
    n = len(s)
    terms = [1/s[i] for i in range(0,n)]
    result = n/sum(terms)
    return result
````

````python
def aritmetic_mean(s):
    "Compute the aritmetic mean of the numbers in the sequence s."
    return sum(s)/len(s)
````

Vamos definir a média geométrica usando uma propriedade de logaritmo para reduzir um produtório em somatório: se $a=x_1 ... x_n$, então

$$a=e^{(ln\, x_1 + ln\, x_2+...+ ln\, x_n)}$$

Logo, se $s=(x_1, ..., x_n)$, então a média geométrica de $s$ é:

$$ (x_1 ... x_n)^{(1/n)} = e^{\dfrac{ln\, x_1 + ln\, x_2+...+ ln\, x_n}{n}} $$

````python
import math
def geometric_mean(s):
    "Compute the geometric mean of the numbers in the sequence s."
    n = len(s)
    # math.log(x) calcula o logaritmo natural de x
    terms = [math.log(s[i]) for i in range(0,n)]
    # math.exp(x) calcula e^x
    result = math.exp(sum(terms)/n)
    return result

def geometric_mean2(s):
    "Compute the geometric mean of the numbers in the sequence s."
    n = len(s)
    # math.prod(s) retorna o produto dos elementos de s
    result = math.prod(s)**(1/n)
    return result
````

Vamos testar nossa função:

````python
harmonic_mean([1,1,1,1])
````

````python
harmonic_mean([1,2,3])
````

````python
aritmetic_mean([1,2,3])
````

````python
geometric_mean([4,9])
````

````python
geometric_mean([1,2,3])
````

````python
geometric_mean2([1,2,3])
````

### Somas de Riemann

Escreva uma função chamada `mn_integral` que receba os parâmetros de entrada `m`, `n`, `a`, `b` e `N` e retorne a [soma de Riemann](https://en.wikipedia.org/wiki/Riemann_sum) à direita:
$$
\int_a^b f(x) \, dx \approx \sum_{k=1}^N f(x_k) \Delta x \ \ , \ \ f(x) = \frac{x^m + 1}{x^n + 1}
$$

e $\Delta x = (b-a)/N$ e $x_k = a + k \Delta x$.

````python
a=0
b=1
N=10
delta_x = (b - a)/N
x = [a + k*delta_x for k in range(0,N+1)]
````

````python
print(x[1])
````

````python
def mn_integral(m,n,a,b,N):
    '''Compute the right Riemann sum for the function f(x) = (x^m + 1)/(x^n + 1)
    on interval [a,b] with a partition of N subintervals of equal size.
    '''
    delta_x = (b - a)/N
    # lista de valores x_k:
    x = [a + k*delta_x for k in range(0,N+1)]
    # lista de valores f(x_k)\Delta_x:
    terms = [(x[k]**m + 1)/(x[k]**n + 1)*delta_x for k in range(1,N+1)]
    riemann_sum = sum(terms)
    return riemann_sum
````

Vamos testar nossa função com uma entrada para a qual conhecemos o resultado. Seja $m=0$, $n=1$, $a=0$, $b=1$ e $N=2$. Então $x_0 = 0$, $x_1 = 1/2$, $x_2 = 1$ e $\Delta x = 1/2$, e calculamos:

$$
\begin{aligned}
\sum_{k=1}^N f(x_k) \Delta x &= \sum_{k=1}^2 \frac{x_k^0 + 1}{x_k^1 + 1} \Delta x \\\
&= \frac{2}{(1/2) + 1} \cdot \frac{1}{2} + \frac{2}{1 + 1} \cdot \frac{1}{2} \\\
&= \frac{7}{6}
\end{aligned}
$$

````python
mn_integral(0,1,0,1,2)
````

````python
7/6
````

Vamos testar nossa função em outro exemplo. Seja $m=1$, $n=2$, $a=0$ e $b=1$. Podemos resolver esta integral exatamente:

$$
\begin{aligned}
\int_0^1 \frac{x + 1}{x^2 + 1} dx &= \int_0^1 \frac{x}{x^2 + 1} dx + \int_0^1 \frac{1}{x^2 + 1} dx \\\
&= \left. \left( \frac{1}{2} \ln(x^2 + 1) + \arctan x \right) \right|_0^1 \\\
&= \frac{1}{2} \ln(2) + \frac{\pi}{4}
\end{aligned}
$$

Aproxime esta integral com uma soma de Riemann para $N=100000$:

````python
mn_integral(1,2,0,1,100000)
````

Como $\pi \approx 3.14159265$ e $\ln(2) \approx 0.69314718$, comparamos com a aproximação:

````python
0.5*math.log(2) + math.pi/4
````

Nossa função calcula os valores esperados!
