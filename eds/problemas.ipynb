### **Problema 1: Método de Runge-Kutta de 4ª Ordem (RK4)**

O método de Euler é simples, mas acumula erros rapidamente. O método de Runge-Kutta de 4ª ordem (RK4) é um dos métodos de passo único mais usados na prática. Ele avalia a função $f(t,y)$ em quatro pontos dentro de cada passo $h$ e combina essas inclinações:

$$y_{n+1} = y_n + \frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4), \qquad t_{n+1} = t_n + h,$$

onde

$$k_1 = f(t_n, y_n), \qquad k_2 = f\left(t_n + \frac{h}{2}, y_n + k_1 \frac{h}{2}\right),$$
$$k_3 = f\left(t_n + \frac{h}{2}, y_n + k_2 \frac{h}{2}\right), \qquad k_4 = f(t_n + h, y_n + k_3 h).$$

**Tarefas:**

1. **Implementação.** Crie uma função `odeRK4(f, t, y0)`, no mesmo formato das funções `odeEuler` e `odeHeun` das notas de aula, que funcione tanto para equações escalares quanto para sistemas (quando `y0` é uma lista ou array).

2. **Interpretação.** Explique o significado de cada $k_i$: em que ponto cada inclinação é calculada? Mostre que, quando $f$ depende apenas de $t$, o RK4 se reduz à regra de Simpson para calcular $\int_{t_n}^{t_{n+1}} f(t)\,dt$.

3. **Ordem do erro global.** Resolva $y' = y$, $y(0)=1$, no intervalo $[0, 5]$, com Euler, Heun e RK4, para $N = 10, 20, 50, 100, 200, 500, 1000$ passos. 4. **Eficiência.** Um passo de Euler avalia $f$ uma vez; um passo de Heun, duas vezes; um passo de RK4, quatro vezes. ### **Problema 2: Modelo Epidemiológico SIR**

O modelo SIR é um dos modelos compartimentais mais fundamentais em epidemiologia. Ele divide a população ($N$) em três grupos:

* **S (Suscetíveis):** indivíduos que podem contrair a doença;
* **I (Infectados):** indivíduos que têm a doença e podem transmiti-la;
* **R (Removidos):** indivíduos que se recuperaram (e ganharam imunidade) ou faleceram.

O modelo é descrito por um sistema de 3 EDOs não lineares:

$$\frac{dS}{dt} = -\frac{\beta S I}{N}, \qquad \frac{dI}{dt} = \frac{\beta S I}{N} - \gamma I, \qquad \frac{dR}{dt} = \gamma I,$$

onde $N = S + I + R$ é a população total, $\beta$ é a taxa de transmissão e $\gamma$ é a taxa de recuperação.

3. **O pico.** Encontre o número máximo de infectados e o dia em que ele ocorre. Usando a equação de $dI/dt$, mostre que no pico $S = N/\mathcal{R}_0$, onde $\mathcal{R}_0 = \beta/\gamma$ é o número básico de reprodução, e confira esse valor na sua simulação.

    * Resolva essa equação com um dos métodos de busca de raízes da primeira parte do curso e compare a fração total de infectados $1 - s_\infty$ com o valor obtido na simulação.

5. **Vacinação.** Suponha que, antes do início da epidemia, uma fração $p$ dos suscetíveis seja vacinada (passando diretamente para o grupo $R$).

### **Problema 3: Equações de Lorenz**

As equações de Lorenz são o sistema de equações diferenciais não lineares

\begin{align}
\frac{dx}{dt} &= \sigma(y - x) \\\
\frac{dy}{dt} &= x(\rho - z) - y \\\
\frac{dz}{dt} &= xy - \beta z
\end{align}

onde $\sigma$, $\rho$ e $\beta$ são números positivos. Elas foram obtidas por Edward Lorenz em 1963 como um modelo muito simplificado da convecção na atmosfera, e são um exemplo clássico de sistema dinâmico caótico.
